```python
from pathlib import Path

path = Path("/usr/local/lib/python3.13/dist-packages/transformers/generation/utils.py")

text = path.read_text()

old = ".ne(eos_token_id_tensor.unsqueeze(1)).prod(dim=0)"
new = ".ne(eos_token_id_tensor.unsqueeze(1)).all(dim=0)"

print("Ketemu:", old in text)

text = text.replace(old, new)

path.write_text(text)

print("Patch selesai.")
```

```python
import requests

image_path = ".//images/immigrants.png"

with open(image_path, "rb") as f:
    response = requests.post(
        "http://127.0.0.1:8000/predict",
        files={
            "image": ("immigrants.png", f, "image/png")
        },
        data={
            "question": "Apa nilai tertinggi pada grafik? jawab menggunakan langkah perhitungannya."
        }
    )

print("Status:", response.status_code)
print("Response:", response.text)
```
