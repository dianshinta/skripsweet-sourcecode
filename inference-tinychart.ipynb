```python
# ============================================================
# STRATIFIED RANDOM SAMPLING 500 NON-PoT
#
# KRITERIA:
#   - Hanya data yang memiliki score
#   - Hanya Non-PoT
#   - Selection dikeluarkan
#
# DESAIN SAMPLING:
#   1. Question Type dibuat seimbang
#   2. Dalam setiap Question Type:
#      Region × Chart Type dialokasikan proporsional
#
# QUESTION TYPE:
#   Sum, Average, Min, Max, Ranking, Compare
#
# OUTPUT:
#   1. CSV 500 sample
#   2. Non-PoT question
#   3. PoT question
#   4. Score Non-PoT
#
# RANDOM SEED = 42
# ============================================================


!pip install -q datasets

import random
import math
import pandas as pd
import json
import glob
import os
import re

from collections import defaultdict
from datasets import load_dataset


# ============================================================
# CONFIG
# ============================================================

DATASET_NAME = "shiinn97/indochartqav2"

SAMPLE_SIZE = 500
RANDOM_SEED = 42

# Kalimat instruksi PoT
POT_PHRASE = "Jawab menggunakan langkah perhitungannya"

# Folder hasil evaluasi TinyChart non-PoT
EVAL_DIR = "/kaggle/input/datasets/dianshinta09/eval-results-ver2"

# Output
OUTPUT_FILE = (
    "/kaggle/working/"
    "indochartqa_500_nonpot_pot_question.csv"
)


# ============================================================
# RANDOM SEED
# ============================================================

random.seed(RANDOM_SEED)


# ============================================================
# FUNGSI AMBIL HUMAN QUESTION
# ============================================================

def get_human_question(conversations):

    if isinstance(conversations, str):

        conversations = json.loads(
            conversations
        )

    for message in conversations:

        if message.get("from") == "human":

            question = message.get(
                "value",
                ""
            )

            question = question.replace(
                "<image>\n",
                ""
            ).strip()

            return question

    return ""


# ============================================================
# LOAD SCORE
# ============================================================

print("=" * 70)
print("MEMBACA SCORE")
print("=" * 70)

jsonl_files = glob.glob(
    os.path.join(
        EVAL_DIR,
        "*.jsonl"
    )
)

print(
    "Jumlah file JSONL:",
    len(jsonl_files)
)


score_dict = {}


for file_path in jsonl_files:

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            try:

                data = json.loads(
                    line
                )

            except json.JSONDecodeError:

                continue

            if (
                "id" in data
                and "score" in data
            ):

                score_dict[
                    str(data["id"])
                ] = data["score"]


print(
    "Jumlah ID yang memiliki score:",
    len(score_dict)
)


# ============================================================
# LOAD DATASET
# ============================================================

print("\n" + "=" * 70)
print("LOAD TEST SET")
print("=" * 70)

dataset = load_dataset(
    DATASET_NAME
)

test_data = dataset["test"]

print(
    "Jumlah test set:",
    len(test_data)
)

print(
    "Kolom dataset:",
    test_data.column_names
)


# ============================================================
# PARSE ID
# ============================================================

pattern = re.compile(
    r"^([BS])\d+(BAR|GBAR|LINE)"
    r"(Sum|Average|Min|Max|Ranking|Compare|Selection)_\d+$"
)


def parse_id(sample_id):

    match = pattern.match(
        sample_id
    )

    if not match:
        return None

    region = match.group(1)

    chart_type = match.group(2)

    question_type = match.group(3)

    return (
        region,
        chart_type,
        question_type
    )


# ============================================================
# FILTER DATA
#
# SYARAT:
#   1. Memiliki score
#   2. ID dapat diparse
#   3. Bukan Selection
#   4. Bukan PoT
# ============================================================

eligible_indices = []

for idx, sample in enumerate(
    test_data
):

    sample_id = str(
        sample["id"]
    )


    # --------------------------------------------------------
    # HARUS MEMILIKI SCORE
    # --------------------------------------------------------

    if sample_id not in score_dict:

        continue


    # --------------------------------------------------------
    # PARSE ID
    # --------------------------------------------------------

    parsed = parse_id(
        sample_id
    )

    if parsed is None:

        continue


    region, chart_type, question_type = parsed


    # --------------------------------------------------------
    # EXCLUDE SELECTION
    # --------------------------------------------------------

    if question_type == "Selection":

        continue


    # --------------------------------------------------------
    # HANYA NON-PoT
    # --------------------------------------------------------

    question = get_human_question(
        sample["conversations"]
    )

    if (
        POT_PHRASE.lower()
        not in question.lower()
    ):

        eligible_indices.append(
            idx
        )


# ============================================================
# INFORMASI ELIGIBLE
# ============================================================

print("\n" + "=" * 70)
print("ELIGIBLE NON-PoT")
print("=" * 70)

print(
    "Non-PoT yang memiliki score:",
    len(eligible_indices)
)


# ============================================================
# VALIDASI SAMPLE SIZE
# ============================================================

if len(eligible_indices) < SAMPLE_SIZE:

    raise ValueError(
        f"Hanya terdapat "
        f"{len(eligible_indices)} "
        f"data eligible. "
        f"Dibutuhkan {SAMPLE_SIZE}."
    )


# ============================================================
# KELOMPOKKAN:
#
# QUESTION TYPE
#       ↓
# REGION × CHART TYPE
# ============================================================

question_groups = defaultdict(
    lambda: defaultdict(list)
)


for idx in eligible_indices:

    sample_id = str(
        test_data[idx]["id"]
    )

    parsed = parse_id(
        sample_id
    )

    if parsed is None:

        continue

    region, chart_type, question_type = parsed

    question_groups[
        question_type
    ][
        (region, chart_type)
    ].append(idx)


# ============================================================
# INFORMASI POPULASI PER QUESTION TYPE
# ============================================================

print("\n" + "=" * 70)
print("DISTRIBUSI POPULASI QUESTION TYPE")
print("=" * 70)


for question_type in sorted(
    question_groups.keys()
):

    total_population = sum(
        len(indices)
        for indices
        in question_groups[
            question_type
        ].values()
    )

    print(
        f"{question_type:<10}: "
        f"{total_population:,}"
    )


# ============================================================
# VALIDASI QUESTION TYPE
# ============================================================

question_types = sorted(
    question_groups.keys()
)


print(
    "\nJumlah Question Type:",
    len(question_types)
)


EXPECTED_QUESTION_TYPES = {
    "Sum",
    "Average",
    "Min",
    "Max",
    "Ranking",
    "Compare"
}


if set(question_types) != EXPECTED_QUESTION_TYPES:

    raise ValueError(
        "Question Type tidak sesuai.\n"
        f"Ditemukan: {question_types}\n"
        f"Diharapkan: "
        f"{sorted(EXPECTED_QUESTION_TYPES)}"
    )


# ============================================================
# ALOKASI SEIMBANG QUESTION TYPE
#
# 500 / 6 = 83 sisa 2
#
# Jadi:
#   4 tipe = 83
#   2 tipe = 84
# ============================================================

n_types = len(
    question_types
)

base_sample = (
    SAMPLE_SIZE
    // n_types
)

remainder = (
    SAMPLE_SIZE
    % n_types
)


question_allocation = {}


for i, question_type in enumerate(
    question_types
):

    n_sample = base_sample

    if i < remainder:

        n_sample += 1

    question_allocation[
        question_type
    ] = n_sample


# ============================================================
# VALIDASI POPULASI QUESTION TYPE
# ============================================================

for question_type in question_types:

    population = sum(
        len(indices)
        for indices
        in question_groups[
            question_type
        ].values()
    )

    required = question_allocation[
        question_type
    ]

    if population < required:

        raise ValueError(
            f"{question_type} hanya memiliki "
            f"{population} data, tetapi "
            f"membutuhkan {required} sample."
        )


# ============================================================
# TAMPILKAN TARGET
# ============================================================

print("\n" + "=" * 70)
print("TARGET SAMPLE PER QUESTION TYPE")
print("=" * 70)


for question_type in question_types:

    print(
        f"{question_type:<10}: "
        f"{question_allocation[question_type]}"
    )


print(
    "\nTotal:",
    sum(
        question_allocation.values()
    )
)


assert (
    sum(
        question_allocation.values()
    )
    == SAMPLE_SIZE
)


# ============================================================
# STRATIFIED RANDOM SAMPLING
#
# QUESTION TYPE → SEIMBANG
# REGION × CHART TYPE → PROPORSIONAL
# ============================================================

selected_indices = []

summary = []


for question_type in question_types:

    groups = question_groups[
        question_type
    ]

    target = question_allocation[
        question_type
    ]


    # --------------------------------------------------------
    # TOTAL POPULASI QUESTION TYPE
    # --------------------------------------------------------

    total_population = sum(
        len(indices)
        for indices
        in groups.values()
    )


    # --------------------------------------------------------
    # ALOKASI REGION × CHART TYPE
    # SECARA PROPORSIONAL
    # --------------------------------------------------------

    allocation = {}


    for group_key, indices in groups.items():

        population = len(
            indices
        )

        ideal = (
            population
            / total_population
            * target
        )

        floor_value = math.floor(
            ideal
        )

        allocation[group_key] = {

            "population":
                population,

            "ideal":
                ideal,

            "sample":
                floor_value,

            "remainder":
                ideal - floor_value
        }


    # --------------------------------------------------------
    # SESUAIKAN AGAR TOTAL SAMA DENGAN TARGET
    # --------------------------------------------------------

    current_total = sum(
        x["sample"]
        for x in allocation.values()
    )

    difference = (
        target
        - current_total
    )


    # --------------------------------------------------------
    # TAMBAH BERDASARKAN REMAINDER TERBESAR
    # --------------------------------------------------------

    candidates = sorted(
        allocation.keys(),
        key=lambda x:
            allocation[x]["remainder"],
        reverse=True
    )


    while difference > 0:

        changed = False

        for key in candidates:

            if difference == 0:

                break


            if (
                allocation[key]["sample"]
                <
                allocation[key]["population"]
            ):

                allocation[key][
                    "sample"
                ] += 1

                difference -= 1

                changed = True


        if not changed:

            break


    # --------------------------------------------------------
    # RANDOM SAMPLING PER REGION × CHART TYPE
    # --------------------------------------------------------

    for group_key in sorted(
        groups.keys()
    ):

        indices = groups[
            group_key
        ]

        n_sample = allocation[
            group_key
        ]["sample"]


        if n_sample == 0:

            continue


        selected = random.sample(
            indices,
            n_sample
        )


        selected_indices.extend(
            selected
        )


        region, chart_type = (
            group_key
        )


        summary.append({

            "Question_Type":
                question_type,

            "Region":
                region,

            "Chart_Type":
                chart_type,

            "Population":
                len(indices),

            "Population_Percentage":
                len(indices)
                / total_population
                * 100,

            "Sample":
                n_sample,

            "Sample_Percentage":
                n_sample
                / target
                * 100
        })


# ============================================================
# VALIDASI TOTAL SAMPLE
# ============================================================

print("\n" + "=" * 70)
print("VALIDASI SAMPLING")
print("=" * 70)


print(
    "Jumlah sample:",
    len(selected_indices)
)


assert (
    len(selected_indices)
    == SAMPLE_SIZE
)


# ============================================================
# ACAK URUTAN SAMPLE
# ============================================================

random.shuffle(
    selected_indices
)


# ============================================================
# AMBIL DATA
# ============================================================

sampled_dataset = test_data.select(
    selected_indices
)

df = sampled_dataset.to_pandas()


# ============================================================
# SIMPAN QUESTION NON-PoT
# ============================================================

df["question_nonpot"] = (
    df["conversations"]
    .apply(
        get_human_question
    )
)


# ============================================================
# BUAT QUESTION PoT
# ============================================================

df["question_pot"] = (
    df["question_nonpot"]
    .astype(str)
    .str.rstrip()
    + " "
    + POT_PHRASE
)


# ============================================================
# TAMBAHKAN SCORE NON-PoT
# ============================================================

df["score_nonpot"] = (
    df["id"]
    .astype(str)
    .map(score_dict)
)


# ============================================================
# VALIDASI FINAL
# ============================================================

print("\n" + "=" * 70)
print("VALIDASI FINAL")
print("=" * 70)


print(
    "Jumlah sample:",
    len(df)
)


print(
    "Score tersedia:",
    df["score_nonpot"]
    .notna()
    .sum()
)


print(
    "Score missing:",
    df["score_nonpot"]
    .isna()
    .sum()
)


assert len(df) == SAMPLE_SIZE

assert (
    df["score_nonpot"]
    .notna()
    .all()
)


# ============================================================
# AVERAGE SCORE NON-PoT
# ============================================================

average_nonpot = (
    df["score_nonpot"]
    .mean()
)


print(
    "\nAverage Score Non-PoT:",
    f"{average_nonpot:.4f}"
)


print(
    "Average Score (%):",
    f"{average_nonpot * 100:.2f}%"
)


# ============================================================
# SUMMARY STRATA
# ============================================================

summary_df = pd.DataFrame(
    summary
)


# ============================================================
# DISTRIBUSI QUESTION TYPE
# ============================================================

print("\n" + "=" * 70)
print("DISTRIBUSI QUESTION TYPE HASIL SAMPLING")
print("=" * 70)


question_summary = (
    summary_df
    .groupby(
        "Question_Type"
    )["Sample"]
    .sum()
    .reset_index()
)


display(
    question_summary
)


# ============================================================
# DISTRIBUSI REGION × CHART TYPE × QUESTION TYPE
# ============================================================

print("\n" + "=" * 70)
print(
    "DISTRIBUSI REGION × CHART TYPE × QUESTION TYPE"
)
print("=" * 70)


display(
    summary_df.sort_values(
        [
            "Question_Type",
            "Region",
            "Chart_Type"
        ]
    )
)


# ============================================================
# CEK CONTOH QUESTION
# ============================================================

print("\n" + "=" * 70)
print("CONTOH QUESTION")
print("=" * 70)


display(
    df[
        [
            "id",
            "question_nonpot",
            "question_pot",
            "score_nonpot"
        ]
    ].head(10)
)


# ============================================================
# SAVE CSV
# ============================================================

df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# SELESAI
# ============================================================

print("\n" + "=" * 70)
print("STRATIFIED SAMPLING SELESAI")
print("=" * 70)


print(
    "Test set awal        :",
    len(test_data)
)


print(
    "Eligible non-PoT     :",
    len(eligible_indices)
)


print(
    "Jumlah Question Type :",
    len(question_types)
)


print(
    "Jumlah sample        :",
    len(df)
)


print(
    "Average Non-PoT      :",
    f"{average_nonpot:.4f}"
)


print(
    "Average Non-PoT (%)  :",
    f"{average_nonpot * 100:.2f}%"
)


print(
    "\nOutput:",
    OUTPUT_FILE
)
```

```python
import json
import re
from pathlib import Path

import pandas as pd
import numpy as np


# ============================================================
# PATH
# ============================================================

CSV_PATH = (
    "/kaggle/input/datasets/"
    "dianshinta09/500-inference-pot-2/"
    "indochartqa_500_pot_inference (2).csv"
)

GROUND_TRUTH_DIR = (
    "/kaggle/input/datasets/"
    "dianshinta09/eval-results-ver2"
)

OUTPUT_DETAIL = (
    "/kaggle/working/"
    "comparison_pot_vs_nonpot_detail.csv"
)


# ============================================================
# SETTINGS
# ============================================================

NUMERIC_TOLERANCE = 0.05
NLS_THRESHOLD = 0.5

YEAR_MIN = 1900
YEAR_MAX = 2026


# ============================================================
# CLEAN TEXT
# ============================================================

def clean_text(value):

    if value is None:
        return ""

    try:
        if pd.isna(value):
            return ""
    except:
        pass

    value = str(value).strip()

    value = re.sub(
        r"\s+",
        " ",
        value
    )

    return value


# ============================================================
# NORMALIZE ID
# ============================================================

def normalize_id(value):

    if value is None:
        return ""

    try:
        if pd.isna(value):
            return ""
    except:
        pass

    value = str(value).strip()

    if re.fullmatch(
        r"-?\d+\.0+",
        value
    ):

        value = value.split(".")[0]

    return value


# ============================================================
# YEAR
# ============================================================

def is_year(
    answer,
    question
):

    answer = clean_text(
        answer
    )

    question = clean_text(
        question
    ).lower()

    if "tahun" not in question:
        return False

    years = re.findall(
        r"\b(19\d{2}|20\d{2})\b",
        answer
    )

    return any(
        YEAR_MIN <= int(year) <= YEAR_MAX
        for year in years
    )


# ============================================================
# NUMERIC
# ============================================================

def numeric_candidates(value):

    value = clean_text(value)

    if not value:
        return []

    value = value.replace(
        "%",
        ""
    )

    value = re.sub(
        r"(?i)\brp\.?",
        "",
        value
    )

    value = value.replace(
        " ",
        ""
    )

    if not re.fullmatch(
        r"-?[\d.,]+",
        value
    ):

        return []

    candidates = []

    def add(x):

        try:

            number = float(x)

            if np.isfinite(number):

                if number not in candidates:

                    candidates.append(
                        number
                    )

        except:
            pass

    comma = value.count(",")
    dot = value.count(".")

    # --------------------------------------------------------
    # Tidak ada separator
    # --------------------------------------------------------

    if comma == 0 and dot == 0:

        add(value)

        return candidates

    # --------------------------------------------------------
    # Ada koma dan titik
    # --------------------------------------------------------

    if comma > 0 and dot > 0:

        if value.rfind(",") > value.rfind("."):

            normalized = (
                value
                .replace(".", "")
                .replace(",", ".")
            )

        else:

            normalized = value.replace(
                ",",
                ""
            )

        add(normalized)

        return candidates

    # --------------------------------------------------------
    # Koma saja
    # --------------------------------------------------------

    if comma > 0:

        parts = value.split(",")

        if len(parts) == 2:

            left, right = parts

            if len(right) == 3:

                add(
                    left + right
                )

                add(
                    left + "." + right
                )

            else:

                add(
                    left + "." + right
                )

        return candidates

    # --------------------------------------------------------
    # Titik saja
    # --------------------------------------------------------

    if dot > 0:

        parts = value.split(".")

        if len(parts) == 2:

            left, right = parts

            if len(right) == 3:

                add(
                    left + right
                )

                add(
                    left + "." + right
                )

            else:

                add(value)

        return candidates

    return candidates


def is_numeric(value):

    return bool(
        numeric_candidates(value)
    )


# ============================================================
# RELAXED ACCURACY
# ============================================================

def numeric_equal(
    prediction,
    ground_truth
):

    predictions = numeric_candidates(
        prediction
    )

    ground_truths = numeric_candidates(
        ground_truth
    )

    if not predictions or not ground_truths:
        return False

    for pred in predictions:

        for gt in ground_truths:

            if gt == 0:

                if pred == 0:
                    return True

            else:

                relative_error = (
                    abs(pred - gt)
                    / abs(gt)
                )

                if relative_error <= NUMERIC_TOLERANCE:

                    return True

    return False


# ============================================================
# LEVENSHTEIN
# ============================================================

def levenshtein(
    s1,
    s2
):

    if s1 == s2:
        return 0

    if not s1:
        return len(s2)

    if not s2:
        return len(s1)

    previous = list(
        range(len(s2) + 1)
    )

    for i, c1 in enumerate(
        s1,
        1
    ):

        current = [i]

        for j, c2 in enumerate(
            s2,
            1
        ):

            current.append(
                min(
                    current[j - 1] + 1,
                    previous[j] + 1,
                    previous[j - 1] + (
                        c1 != c2
                    )
                )
            )

        previous = current

    return previous[-1]


# ============================================================
# NLS
# ============================================================

def compute_nls(
    ground_truth,
    prediction
):

    ground_truth = clean_text(
        ground_truth
    ).lower()

    prediction = clean_text(
        prediction
    ).lower()

    if ground_truth == prediction:
        return 1.0

    if not ground_truth or not prediction:
        return 0.0

    distance = levenshtein(
        ground_truth,
        prediction
    )

    max_length = max(
        len(ground_truth),
        len(prediction)
    )

    return 1 - (
        distance / max_length
    )


# ============================================================
# EM
# ============================================================

def compute_em(
    prediction,
    ground_truth
):

    return float(
        clean_text(prediction)
        == clean_text(ground_truth)
    )


# ============================================================
# EVALUATE PoT
# ============================================================

def evaluate_pot(
    prediction,
    ground_truth,
    question
):

    prediction = clean_text(
        prediction
    )

    ground_truth = clean_text(
        ground_truth
    )

    # --------------------------------------------------------
    # YEAR
    # --------------------------------------------------------

    if is_year(
        prediction,
        question
    ):

        em = compute_em(
            prediction,
            ground_truth
        )

        return {
            "score": em,
            "metric": "EM",
            "answer_type": "year",
            "em": em,
            "nls": compute_nls(
                ground_truth,
                prediction
            ),
            "relaxed_acc": None
        }

    # --------------------------------------------------------
    # NUMERIC
    # --------------------------------------------------------

    if is_numeric(
        prediction
    ):

        relaxed_acc = float(
            numeric_equal(
                prediction,
                ground_truth
            )
        )

        return {
            "score": relaxed_acc,
            "metric": "Relaxed Accuracy",
            "answer_type": "numeric",
            "em": compute_em(
                prediction,
                ground_truth
            ),
            "nls": compute_nls(
                ground_truth,
                prediction
            ),
            "relaxed_acc": relaxed_acc
        }

    # --------------------------------------------------------
    # TEXT
    # --------------------------------------------------------

    nls = compute_nls(
        ground_truth,
        prediction
    )

    return {
        "score": nls,
        "metric": "NLS without threshold",
        "answer_type": "text",
        "em": compute_em(
            prediction,
            ground_truth
        ),
        "nls": nls,
        "relaxed_acc": None
    }


# ============================================================
# LOAD GROUND TRUTH
# ============================================================

def load_ground_truth(
    directory
):

    directory = Path(
        directory
    )

    gt_by_id = {}

    jsonl_files = sorted(
        directory.glob(
            "*.jsonl"
        )
    )

    print(
        f"Jumlah file GT: "
        f"{len(jsonl_files)}"
    )

    for file in jsonl_files:

        with open(
            file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                if not line.strip():
                    continue

                try:

                    record = json.loads(
                        line
                    )

                except json.JSONDecodeError:

                    continue

                record_id = normalize_id(
                    record.get(
                        "id",
                        ""
                    )
                )

                if not record_id:

                    record_id = normalize_id(
                        record.get(
                            "ID",
                            ""
                        )
                    )

                if not record_id:
                    continue

                gt_by_id[
                    record_id
                ] = {

                    "question":
                        record.get(
                            "question",
                            ""
                        ),

                    "gt_answer":
                        record.get(
                            "gt_answer",
                            record.get(
                                "answer",
                                ""
                            )
                        ),

                    "chart_type":
                        record.get(
                            "chart_type",
                            ""
                        ),

                    "question_type":
                        record.get(
                            "question_type",
                            ""
                        )
                }

    print(
        f"Jumlah GT unik: "
        f"{len(gt_by_id):,}"
    )

    return gt_by_id


# ============================================================
# MAIN COMPARISON
# ============================================================

def main():

    # --------------------------------------------------------
    # LOAD CSV
    # --------------------------------------------------------

    df = pd.read_csv(
        CSV_PATH,
        dtype=str,
        keep_default_na=False
    )

    print(
        "=" * 80
    )

    print(
        "CSV"
    )

    print(
        "=" * 80
    )

    print(
        f"Jumlah data: {len(df):,}"
    )

    print(
        f"Kolom: {list(df.columns)}"
    )

    # --------------------------------------------------------
    # CHECK COLUMN
    # --------------------------------------------------------

    required = [
        "id",
        "answer_pot",
        "score_nonpot"
    ]

    missing = [
        col
        for col in required
        if col not in df.columns
    ]

    if missing:

        raise ValueError(
            f"Kolom tidak ditemukan: "
            f"{missing}"
        )

    # --------------------------------------------------------
    # LOAD GT
    # --------------------------------------------------------

    gt_by_id = load_ground_truth(
        GROUND_TRUTH_DIR
    )

    # --------------------------------------------------------
    # EVALUATION
    # --------------------------------------------------------

    records = []

    missing_ids = []

    for _, row in df.iterrows():

        record_id = normalize_id(
            row["id"]
        )

        # ----------------------------------------------------
        # Find GT
        # ----------------------------------------------------

        if record_id not in gt_by_id:

            missing_ids.append(
                record_id
            )

            continue

        gt = gt_by_id[
            record_id
        ]

        question = clean_text(
            gt["question"]
        )

        gt_answer = clean_text(
            gt["gt_answer"]
        )

        # ----------------------------------------------------
        # PoT
        # ----------------------------------------------------

        answer_pot = clean_text(
            row["answer_pot"]
        )

        pot = evaluate_pot(
            answer_pot,
            gt_answer,
            question
        )

        # ----------------------------------------------------
        # Non-PoT
        # ----------------------------------------------------
        # LANGSUNG AMBIL DARI CSV
        # Tidak dihitung ulang.
        # ----------------------------------------------------

        score_nonpot = pd.to_numeric(
            row["score_nonpot"],
            errors="coerce"
        )

        # ----------------------------------------------------
        # Difference
        # ----------------------------------------------------

        difference = (
            pot["score"]
            - score_nonpot
            if pd.notna(score_nonpot)
            else np.nan
        )

        records.append({

            "id":
                record_id,

            "chart_type":
                gt["chart_type"],

            "question_type":
                gt["question_type"],

            "question":
                question,

            "gt_answer":
                gt_answer,

            "answer_pot":
                answer_pot,

            "score_pot":
                pot["score"],

            "pot_metric":
                pot["metric"],

            "pot_answer_type":
                pot["answer_type"],

            "pot_em":
                pot["em"],

            "pot_nls":
                pot["nls"],

            "pot_relaxed_acc":
                pot["relaxed_acc"],

            "score_nonpot":
                score_nonpot,

            "difference_pot_minus_nonpot":
                difference
        })

    # --------------------------------------------------------
    # DETAIL DATAFRAME
    # --------------------------------------------------------

    result = pd.DataFrame(
        records
    )

    # --------------------------------------------------------
    # SAVE DETAIL
    # --------------------------------------------------------

    result.to_csv(
        OUTPUT_DETAIL,
        index=False,
        encoding="utf-8-sig"
    )

    # ========================================================
    # OVERALL
    # ========================================================

    print(
        "\n"
        + "=" * 80
    )

    print(
        "PERBANDINGAN OVERALL"
    )

    print(
        "=" * 80
    )

    print(
        f"CSV rows       : {len(df):,}"
    )

    print(
        f"Berhasil eval  : {len(result):,}"
    )

    print(
        f"GT tidak ada   : {len(missing_ids):,}"
    )

    print(
        f"\nPoT Score      : "
        f"{result['score_pot'].mean():.4f}"
    )

    print(
        f"Non-PoT Score  : "
        f"{result['score_nonpot'].mean():.4f}"
    )

    print(
        f"Difference     : "
        f"{result['difference_pot_minus_nonpot'].mean():+.4f}"
    )

    # ========================================================
    # COUNT
    # ========================================================

    pot_mean = result[
        "score_pot"
    ].mean()

    nonpot_mean = result[
        "score_nonpot"
    ].mean()

    pot_higher = (
        result[
            "score_pot"
        ]
        > result[
            "score_nonpot"
        ]
    ).sum()

    nonpot_higher = (
        result[
            "score_nonpot"
        ]
        > result[
            "score_pot"
        ]
    ).sum()

    equal = (
        result[
            "score_pot"
        ]
        == result[
            "score_nonpot"
        ]
    ).sum()

    print(
        "\n"
        + "=" * 80
    )

    print(
        "JUMLAH HASIL"
    )

    print(
        "=" * 80
    )

    print(
        f"PoT > Non-PoT : {pot_higher}"
    )

    print(
        f"PoT < Non-PoT : {nonpot_higher}"
    )

    print(
        f"PoT = Non-PoT : {equal}"
    )

    # ========================================================
    # BY QUESTION TYPE
    # ========================================================

    print(
        "\n"
        + "=" * 80
    )

    print(
        "PER QUESTION TYPE"
    )

    print(
        "=" * 80
    )

    by_question = (
        result
        .groupby("question_type")
        .agg(
            n=("id", "count"),
            pot_score=(
                "score_pot",
                "mean"
            ),
            nonpot_score=(
                "score_nonpot",
                "mean"
            ),
            difference=(
                "difference_pot_minus_nonpot",
                "mean"
            )
        )
        .reset_index()
    )

    print(
        by_question.to_string(
            index=False
        )
    )

    # ========================================================
    # BY CHART TYPE
    # ========================================================

    print(
        "\n"
        + "=" * 80
    )

    print(
        "PER CHART TYPE"
    )

    print(
        "=" * 80
    )

    by_chart = (
        result
        .groupby("chart_type")
        .agg(
            n=("id", "count"),
            pot_score=(
                "score_pot",
                "mean"
            ),
            nonpot_score=(
                "score_nonpot",
                "mean"
            ),
            difference=(
                "difference_pot_minus_nonpot",
                "mean"
            )
        )
        .reset_index()
    )

    print(
        by_chart.to_string(
            index=False
        )
    )

    # ========================================================
    # MISSING IDS
    # ========================================================

    if missing_ids:

        missing_path = (
            "/kaggle/working/"
            "missing_gt_ids.csv"
        )

        pd.DataFrame({
            "id":
                missing_ids
        }).to_csv(
            missing_path,
            index=False
        )

        print(
            f"\n⚠️ Missing IDs disimpan:"
        )

        print(
            missing_path
        )

    print(
        f"\n✓ Detail hasil:"
    )

    print(
        OUTPUT_DETAIL
    )

    return result


# ============================================================
# RUN
# ============================================================

result = main()
```

```python
import json
import re
import os
import zipfile
from pathlib import Path
from collections import defaultdict, Counter

import pandas as pd
import numpy as np


# ============================================================
# SETTINGS
# ============================================================

NUMERIC_TOLERANCE = 0.05
NLS_THRESHOLD = 0.5

YEAR_MIN = 1900
YEAR_MAX = 2026


# ============================================================
# TEXT CLEANING
# ============================================================

def clean_answer_text(answer):
    """
    Clean answer before classification/evaluation.
    """

    if answer is None:
        return ""

    answer = str(answer).strip()

    # Normalize whitespace
    answer = re.sub(r'\s+', ' ', answer)

    return answer


# ============================================================
# YEAR DETECTION
# ============================================================

def is_year(answer_str, question_str):
    """
    Detect whether the answer represents a year.

    Criteria:
    - Question contains the word 'tahun'
    - Answer contains a 4-digit number between 1900 and 2026
    """

    answer_str = clean_answer_text(answer_str)
    question_str = clean_answer_text(question_str).lower()

    if 'tahun' not in question_str:
        return False

    matches = re.findall(
        r'\b(19\d{2}|20\d{2})\b',
        answer_str
    )

    if not matches:
        return False

    for match in matches:

        year = int(match)

        if YEAR_MIN <= year <= YEAR_MAX:
            return True

    return False


# ============================================================
# NUMERIC PARSER
# ============================================================

def numeric_candidates(value):
    """
    Return possible numeric interpretations of an answer.

    Designed for Indonesian and international number formats.
    """

    if value is None:
        return []

    s = clean_answer_text(value)

    if not s:
        return []

    # --------------------------------------------------------
    # REMOVE COMMON SYMBOLS
    # --------------------------------------------------------

    s = s.replace('%', '')

    s = re.sub(
        r'(?i)\brp\.?',
        '',
        s
    )

    s = s.strip()
    s = s.replace(' ', '')

    # --------------------------------------------------------
    # VALID CHARACTER CHECK
    # --------------------------------------------------------

    if not re.fullmatch(
        r'-?[\d.,]+',
        s
    ):
        return []

    candidates = []

    def add_candidate(value):

        try:

            number = float(value)

            if np.isfinite(number):

                if not any(
                    np.isclose(
                        number,
                        existing,
                        rtol=0,
                        atol=1e-12
                    )
                    for existing in candidates
                ):
                    candidates.append(number)

        except (
            ValueError,
            TypeError
        ):
            pass

    comma_count = s.count(',')
    dot_count = s.count('.')

    # ========================================================
    # NO SEPARATOR
    # ========================================================

    if comma_count == 0 and dot_count == 0:

        add_candidate(s)

        return candidates

    # ========================================================
    # BOTH COMMA AND DOT
    # ========================================================

    if comma_count > 0 and dot_count > 0:

        if s.rfind(',') > s.rfind('.'):

            normalized = (
                s
                .replace('.', '')
                .replace(',', '.')
            )

        else:

            normalized = (
                s
                .replace(',', '')
            )

        add_candidate(normalized)

        return candidates

    # ========================================================
    # COMMA ONLY
    # ========================================================

    if comma_count > 0:

        if comma_count > 1:

            if re.fullmatch(
                r'-?\d{1,3}(,\d{3})+',
                s
            ):

                add_candidate(
                    s.replace(',', '')
                )

            return candidates

        left, right = s.split(',')

        if len(right) != 3:

            add_candidate(
                left + '.' + right
            )

            return candidates

        if (
            len(right) == 3
            and len(left.lstrip('-')) <= 3
        ):

            add_candidate(
                left + '.' + right
            )

            add_candidate(
                left + right
            )

        else:

            add_candidate(
                left + '.' + right
            )

        return candidates

    # ========================================================
    # DOT ONLY
    # ========================================================

    if dot_count > 0:

        if dot_count > 1:

            if re.fullmatch(
                r'-?\d{1,3}(\.\d{3})+',
                s
            ):

                add_candidate(
                    s.replace('.', '')
                )

            return candidates

        left, right = s.split('.')

        if len(right) != 3:

            add_candidate(s)

            return candidates

        if (
            len(right) == 3
            and len(left.lstrip('-')) <= 3
        ):

            add_candidate(
                left + right
            )

            add_candidate(s)

        else:

            add_candidate(s)

        return candidates

    return candidates


# ============================================================
# NUMERIC DETECTION
# ============================================================

def is_numeric(answer_str):

    return len(
        numeric_candidates(answer_str)
    ) > 0


# ============================================================
# NUMERIC COMPARISON
# ============================================================

def numeric_equal(
    model_answer,
    gt_answer,
    tolerance=NUMERIC_TOLERANCE
):

    model_candidates = numeric_candidates(
        model_answer
    )

    gt_candidates = numeric_candidates(
        gt_answer
    )

    if not model_candidates or not gt_candidates:
        return False

    for model_num in model_candidates:

        for gt_num in gt_candidates:

            if gt_num == 0:

                if model_num == 0:
                    return True

                continue

            error_rate = (
                abs(model_num - gt_num)
                / abs(gt_num)
            )

            if error_rate <= tolerance:
                return True

    return False


# ============================================================
# ANSWER TYPE
# ============================================================

def get_answer_type(
    model_answer,
    gt_answer,
    question
):
    """
    Determine answer type based on MODEL ANSWER.

    Priority:
    1. Year
    2. Numeric
    3. Text
    """

    # --------------------------------------------------------
    # YEAR
    # --------------------------------------------------------

    if is_year(
        model_answer,
        question
    ):
        return 'year'

    # --------------------------------------------------------
    # NUMERIC
    # --------------------------------------------------------

    if is_numeric(
        model_answer
    ):
        return 'numeric'

    # --------------------------------------------------------
    # TEXT
    # --------------------------------------------------------

    return 'text'


# ============================================================
# LEVENSHTEIN DISTANCE
# ============================================================

def levenshtein_distance(s1, s2):

    if s1 == s2:
        return 0

    if len(s1) == 0:
        return len(s2)

    if len(s2) == 0:
        return len(s1)

    # Pastikan s2 lebih pendek
    if len(s1) < len(s2):
        s1, s2 = s2, s1

    previous_row = list(
        range(len(s2) + 1)
    )

    for i, c1 in enumerate(
        s1,
        start=1
    ):

        current_row = [i]

        for j, c2 in enumerate(
            s2,
            start=1
        ):

            insertion = (
                current_row[j - 1] + 1
            )

            deletion = (
                previous_row[j] + 1
            )

            substitution = (
                previous_row[j - 1]
                + (c1 != c2)
            )

            current_row.append(
                min(
                    insertion,
                    deletion,
                    substitution
                )
            )

        previous_row = current_row

    return previous_row[-1]


# ============================================================
# NORMALIZED LEVENSHTEIN SIMILARITY (NLS)
# ============================================================

def compute_nls(gt, pred):
    """
    Normalized Levenshtein Similarity (NLS) tanpa threshold.
    Nilai berada pada rentang 0–1.
    """

    gt = clean_answer_text(gt).lower()
    pred = clean_answer_text(pred).lower()

    if gt == pred:
        return 1.0

    if not gt or not pred:
        return 0.0

    distance = levenshtein_distance(
        gt,
        pred
    )

    max_len = max(
        len(gt),
        len(pred)
    )

    return 1.0 - (
        distance / max_len
    )


def compute_nls_with_threshold(
    gt,
    pred,
    threshold=NLS_THRESHOLD
):
    """
    NLS dengan threshold.

    Jika NLS >= threshold,
    nilai NLS dipertahankan.

    Jika NLS < threshold,
    skor menjadi 0.
    """

    nls = compute_nls(
        gt,
        pred
    )

    if nls >= threshold:
        return nls

    return 0.0


# ============================================================
# EXACT MATCH
# ============================================================

def compute_em(
    model_answer,
    gt_answer
):

    model_str = clean_answer_text(
        model_answer
    )

    gt_str = clean_answer_text(
        gt_answer
    )

    return (
        1.0
        if model_str == gt_str
        else 0.0
    )


# ============================================================
# RELAXED ACCURACY
# ============================================================

def compute_relaxed_accuracy(
    model_val,
    gt_val,
    tolerance=NUMERIC_TOLERANCE
):

    return (
        1.0
        if numeric_equal(
            model_val,
            gt_val,
            tolerance
        )
        else 0.0
    )


# ============================================================
# SCORE SAMPLE
# ============================================================

def score_sample(
    model_answer,
    gt_answer,
    question,
    question_type
):
    """
    Evaluation logic:

    Numeric:
        Score = Relaxed Accuracy

    Year:
        Score = EM
        NLS with threshold dan NLS without threshold
        dihitung sebagai metrik tambahan.

    Text:
        Score = NLS without threshold
        EM dan NLS with threshold
        dihitung sebagai metrik tambahan.

    Answer type ditentukan berdasarkan MODEL ANSWER.

    Parameter question_type diterima untuk menjaga konteks
    question type, tetapi tidak memaksa answer type tertentu.
    """

    # ========================================================
    # DETERMINE TYPE
    # ========================================================

    answer_type = get_answer_type(
        model_answer,
        gt_answer,
        question
    )

    # ========================================================
    # NUMERIC
    # ========================================================

    if answer_type == 'numeric':

        relaxed_acc = compute_relaxed_accuracy(
            model_answer,
            gt_answer
        )

        return {
            'score': relaxed_acc,
            'answer_type': 'numeric',
            'metric_used': 'Relaxed Accuracy',

            'em': None,
            'nls_with_threshold': None,
            'nls_without_threshold': None,
            'relaxed_acc': relaxed_acc
        }

    # ========================================================
    # YEAR
    # ========================================================

    elif answer_type == 'year':

        # ----------------------------------------------------
        # EM
        # ----------------------------------------------------

        em = compute_em(
            model_answer,
            gt_answer
        )

        # ----------------------------------------------------
        # NLS WITHOUT THRESHOLD
        # ----------------------------------------------------

        nls_without_threshold = compute_nls(
            model_answer,
            gt_answer
        )

        # ----------------------------------------------------
        # NLS WITH THRESHOLD
        # ----------------------------------------------------

        nls_with_threshold = compute_nls_with_threshold(
            model_answer,
            gt_answer
        )

        # Score utama YEAR = EM
        score = em

        return {
            'score': score,
            'answer_type': 'year',
            'metric_used': 'EM',

            'em': em,
            'nls_with_threshold': nls_with_threshold,
            'nls_without_threshold': nls_without_threshold,
            'relaxed_acc': None
        }

    # ========================================================
    # TEXT
    # ========================================================

    else:

        em = compute_em(
            model_answer,
            gt_answer
        )

        nls_without_threshold = compute_nls(
            model_answer,
            gt_answer
        )

        nls_with_threshold = compute_nls_with_threshold(
            model_answer,
            gt_answer
        )

        # Score utama TEXT = NLS without threshold
        score = nls_without_threshold

        return {
            'score': score,
            'answer_type': 'text',
            'metric_used': 'NLS without threshold',

            'em': em,
            'nls_without_threshold': nls_without_threshold,
            'nls_with_threshold': nls_with_threshold,
            'relaxed_acc': None
        }


# ============================================================
# PARSE FILENAME
# ============================================================

def parse_filename(filename):

    name = filename.replace(
        '.jsonl',
        ''
    )

    pattern = (
        r'([A-Z]{2,})'
        r'(Sum|Average|Min|Max|Ranking|Compare|Selection)$'
    )

    match = re.search(
        pattern,
        name
    )

    if match:

        chart_type = match.group(1)
        question_type = match.group(2)

        return (
            chart_type,
            question_type
        )

    return None, None


# ============================================================
# RECLASSIFY QUESTION TYPE
# ============================================================

def classify_question_type(
    model_answer,
    question,
    original_question_type
):
    """
    Menentukan question type akhir berdasarkan model answer
    dan keyword pada pertanyaan.

    Reklasifikasi hanya dilakukan untuk Sum dan Average.

    Ketentuan:
    - Jika model answer numeric:
        tetap menggunakan original question type.
    - Jika model answer nonnumeric:
        cek keyword pertanyaan.
    - Tipe selain Sum/Average:
        tetap menggunakan original question type.
    """

    # --------------------------------------------------------
    # Hanya Sum dan Average yang dapat direklasifikasi
    # --------------------------------------------------------

    if original_question_type not in [
        "Sum",
        "Average"
    ]:
        return original_question_type

    # --------------------------------------------------------
    # Jika model answer numeric,
    # tetap menggunakan tipe pertanyaan asli
    # --------------------------------------------------------

    if is_numeric(model_answer):
        return original_question_type

    # --------------------------------------------------------
    # Model answer nonnumeric
    # → cek keyword pertanyaan
    # --------------------------------------------------------

    question_lower = clean_answer_text(
        question
    ).lower()

    # --------------------------------------------------------
    # MIN
    # --------------------------------------------------------

    if "terendah" or "terkecil" in question_lower:
        return "Min"

    # --------------------------------------------------------
    # MAX
    # --------------------------------------------------------

    if "tertinggi" or "paling banyak" or "terbesar" in question_lower:
        return "Max"

    # --------------------------------------------------------
    # COMPARE
    # --------------------------------------------------------

    compare_keywords = [
        "lebih tinggi",
        "lebih kecil",
        "lebih banyak",
        "lebih sedikit",
        "lebih besar",
        "lebih rendah",
        "daripada",
        "lebih dari"
    ]

    if any(
        keyword in question_lower
        for keyword in compare_keywords
    ):
        return "Compare"

    # --------------------------------------------------------
    # Jika tidak memenuhi aturan apa pun,
    # tetap gunakan tipe asli
    # --------------------------------------------------------

    return original_question_type


# ============================================================
# EVALUATE PoT COMMANDS
# ============================================================

def evaluate_cmds(cmds):

    env = {
        "np": np
    }

    for cmd in cmds:

        exec(
            cmd,
            env
        )

    answer = env["Answer"]

    # ========================================================
    # LIST / ARRAY WITH ONE ELEMENT
    # ========================================================

    if (
        isinstance(
            answer,
            (
                list,
                np.ndarray
            )
        )
        and len(answer) == 1
    ):

        answer = answer[0]

    # ========================================================
    # LIST / ARRAY WITH MULTIPLE ELEMENTS
    # ========================================================

    if isinstance(
        answer,
        (
            list,
            np.ndarray
        )
    ):

        answer = [
            str(x)
            for x in answer
        ]

        if len(answer) == 0:
            return ""

        if len(answer) == 1:
            return answer[0]

        if len(answer) == 2:

            return (
                answer[0]
                + ' and '
                + answer[1]
            )

        new_answer = answer[0]

        for i in range(
            1,
            len(answer) - 1
        ):

            new_answer += (
                ', '
                + answer[i]
            )

        new_answer += (
            ' and '
            + answer[-1]
        )

        answer = new_answer

    # ========================================================
    # BOOLEAN
    # ========================================================

    if isinstance(
        answer,
        (
            bool,
            np.bool_
        )
    ):

        answer = (
            "Yes"
            if answer
            else "No"
        )

    return answer


# ============================================================
# RESOLVE ANSWER
# ============================================================

def resolve_answer(answer):

    answer = clean_answer_text(
        answer
    )

    # ========================================================
    # CHECK PoT
    # ========================================================

    if (
        "<step>" in answer
        or "<comment>" in answer
        or "Answer =" in answer
    ):

        original_answer = answer

        try:

            # ------------------------------------------------
            # Remove comments
            # ------------------------------------------------

            answer = re.sub(
                r"<comment>.*?</comment>",
                "",
                answer,
                flags=re.DOTALL
            )

            # ------------------------------------------------
            # Remove step tags
            # ------------------------------------------------

            answer = answer.replace(
                "<step>",
                ""
            )

            answer = answer.replace(
                "</step>",
                "\n"
            )

            # ------------------------------------------------
            # Split commands
            # ------------------------------------------------

            cmds = [
                cmd.strip()
                for cmd in answer.split("\n")
                if cmd.strip()
            ]

            # ------------------------------------------------
            # Evaluate
            # ------------------------------------------------

            result = evaluate_cmds(
                cmds
            )

            return clean_answer_text(
                result
            )

        except Exception:

            return original_answer

    # ========================================================
    # NON-PoT
    # ========================================================

    return answer


# ============================================================
# NORMALIZE BOOLEAN
# ============================================================

def normalize_boolean_answer(answer):

    answer = clean_answer_text(
        answer
    )

    mapping = {

        "True": "Ya",
        "False": "Tidak",

        "true": "Ya",
        "false": "Tidak",

        "TRUE": "Ya",
        "FALSE": "Tidak",

        "Yes": "Ya",
        "No": "Tidak",

        "yes": "Ya",
        "no": "Tidak",

        "YES": "Ya",
        "NO": "Tidak"
    }

    return mapping.get(
        answer,
        answer
    )


# ============================================================
# EVALUATE JSONL FILES
# ============================================================

def evaluate_jsonl_files(directory):

    results = defaultdict(
        lambda: {
            'count': 0,
            'scores': [],

            # ------------------------------------------------
            # EM, NLS hanya untuk
            # text + year
            # ------------------------------------------------

            'em_scores': [],
            'nls_without_threshold_scores': [],
            'nls_with_threshold_scores': [],
            'relaxed_acc_scores': [],

            # ------------------------------------------------
            # Jumlah sampel yang diukur
            # oleh masing-masing metrik
            # ------------------------------------------------

            'em_count': 0,
            'nls_without_threshold_count': 0,
            'nls_with_threshold_count': 0,
            'relaxed_acc_count': 0
        }
    )

    ranking_answer_types = Counter()
    ranking_numeric_examples = []

    average_answer_types = Counter()
    average_non_numeric_examples = []

    # ========================================================
    # FIND JSONL
    # ========================================================

    jsonl_files = sorted(
        Path(directory).glob(
            '*.jsonl'
        )
    )

    if not jsonl_files:

        print(
            f"No JSONL files found in {directory}"
        )

        return None

    print(
        f"Found {len(jsonl_files)} JSONL files"
    )

    # ========================================================
    # RESOLVED DIRECTORY
    # ========================================================

    resolved_directory = Path(
        '/kaggle/working/resolved_jsonl'
    )

    resolved_directory.mkdir(
        parents=True,
        exist_ok=True
    )

    total_records = 0

    # ========================================================
    # PROCESS FILES
    # ========================================================

    for jsonl_file in jsonl_files:

        chart_type, original_question_type = parse_filename(
            jsonl_file.name
        )

        if (
            not chart_type
            or not original_question_type
        ):

            print(
                f"⚠️ Skipping "
                f"{jsonl_file.name} "
                f"- couldn't parse filename"
            )

            continue

        resolved_file = (
            resolved_directory
            / jsonl_file.name
        )

        try:

            with open(
                jsonl_file,
                'r',
                encoding='utf-8'
            ) as f, open(
                resolved_file,
                'w',
                encoding='utf-8'
            ) as out:

                for line in f:

                    if not line.strip():
                        continue

                    record = json.loads(
                        line
                    )

                    # =================================================
                    # ORIGINAL VALUES
                    # =================================================

                    model_answer = record.get(
                        'model_answer',
                        ''
                    )

                    gt_answer = record.get(
                        'gt_answer',
                        ''
                    )

                    question = record.get(
                        'question',
                        ''
                    )

                    # =================================================
                    # RESOLVE PoT
                    # =================================================

                    model_answer = resolve_answer(
                        model_answer
                    )

                    gt_answer = resolve_answer(
                        gt_answer
                    )

                    # =================================================
                    # NORMALIZE BOOLEAN
                    # =================================================

                    model_answer = normalize_boolean_answer(
                        model_answer
                    )

                    gt_answer = normalize_boolean_answer(
                        gt_answer
                    )

                    # =================================================
                    # FINAL QUESTION TYPE
                    # =================================================

                    final_question_type = (
                        classify_question_type(
                            model_answer,
                            question,
                            original_question_type
                        )
                    )

                    # =================================================
                    # GROUP KEY
                    #
                    # IMPORTANT:
                    # Menggunakan FINAL question type.
                    # =================================================

                    group_key = (
                        chart_type,
                        final_question_type
                    )

                    # =================================================
                    # UPDATE ANSWERS
                    # =================================================

                    record['model_answer'] = (
                        model_answer
                    )

                    record['gt_answer'] = (
                        gt_answer
                    )

                    record[
                        'original_question_type'
                    ] = (
                        original_question_type
                    )

                    record[
                        'question_type'
                    ] = (
                        final_question_type
                    )

                    # =================================================
                    # SCORE
                    # =================================================

                    metrics = score_sample(
                        model_answer,
                        gt_answer,
                        question,
                        final_question_type
                    )

                    # =================================================
                    # AVERAGE ANSWER TYPE CHECK
                    # =================================================

                    if original_question_type == "Average":

                        answer_type = metrics[
                            'answer_type'
                        ]

                        average_answer_types[
                            answer_type
                        ] += 1

                        if answer_type != "numeric":

                            average_non_numeric_examples.append({
                                "id": record.get(
                                    "id",
                                    ""
                                ),
                                "original_question_type":
                                    original_question_type,
                                "final_question_type":
                                    final_question_type,
                                "question":
                                    question,
                                "gt_answer":
                                    gt_answer,
                                "model_answer":
                                    model_answer,
                                "answer_type":
                                    answer_type,
                                "score":
                                    metrics['score'],
                                "metric_used":
                                    metrics['metric_used']
                            })

                    # =================================================
                    # SAVE METRICS
                    # =================================================

                    record['score'] = (
                        metrics['score']
                    )

                    record['answer_type'] = (
                        metrics['answer_type']
                    )

                    record['metric_used'] = (
                        metrics['metric_used']
                    )

                    record['relaxed_acc'] = (
                        metrics['relaxed_acc']
                    )

                    record['em'] = (
                        metrics['em']
                    )

                    record[
                        'nls_without_threshold'
                    ] = (
                        metrics[
                            'nls_without_threshold'
                        ]
                    )

                    record[
                        'nls_with_threshold'
                    ] = (
                        metrics[
                            'nls_with_threshold'
                        ]
                    )

                    # =================================================
                    # RANKING CHECK
                    # =================================================

                    if original_question_type == "Ranking":

                        answer_type = get_answer_type(
                            model_answer,
                            gt_answer,
                            question
                        )

                        ranking_answer_types[
                            answer_type
                        ] += 1

                        if answer_type == "numeric":

                            ranking_numeric_examples.append({
                                "question":
                                    question,
                                "gt_answer":
                                    gt_answer,
                                "model_answer":
                                    model_answer
                            })

                    # =================================================
                    # SAVE RESOLVED + EVALUATED RECORD
                    # =================================================

                    out.write(
                        json.dumps(
                            record,
                            ensure_ascii=False
                        )
                        + '\n'
                    )

                    # =================================================
                    # UPDATE AGGREGATE
                    # =================================================

                    results[group_key]['count'] += 1

                    results[group_key]['scores'].append(
                        metrics['score']
                    )

                    # -------------------------------------------------
                    # EM
                    # -------------------------------------------------

                    if metrics['em'] is not None:

                        results[group_key][
                            'em_scores'
                        ].append(
                            metrics['em']
                        )

                        results[group_key][
                            'em_count'
                        ] += 1

                    # -------------------------------------------------
                    # NLS WITHOUT THRESHOLD
                    # -------------------------------------------------

                    if (
                        metrics[
                            'nls_without_threshold'
                        ] is not None
                    ):

                        results[group_key][
                            'nls_without_threshold_scores'
                        ].append(
                            metrics[
                                'nls_without_threshold'
                            ]
                        )

                        results[group_key][
                            'nls_without_threshold_count'
                        ] += 1

                    # -------------------------------------------------
                    # NLS WITH THRESHOLD
                    # -------------------------------------------------

                    if (
                        metrics[
                            'nls_with_threshold'
                        ] is not None
                    ):

                        results[group_key][
                            'nls_with_threshold_scores'
                        ].append(
                            metrics[
                                'nls_with_threshold'
                            ]
                        )

                        results[group_key][
                            'nls_with_threshold_count'
                        ] += 1

                    # -------------------------------------------------
                    # RELAXED ACCURACY
                    # -------------------------------------------------

                    if metrics['relaxed_acc'] is not None:

                        results[group_key][
                            'relaxed_acc_scores'
                        ].append(
                            metrics[
                                'relaxed_acc'
                            ]
                        )

                        results[group_key][
                            'relaxed_acc_count'
                        ] += 1

                    total_records += 1

        except Exception as e:

            print(
                f"ERROR processing "
                f"{jsonl_file.name}: {e}"
            )

            continue

    # ========================================================
    # AGGREGATE
    # ========================================================

    data = []

    for (
        chart_type,
        question_type
    ), metrics_dict in sorted(
        results.items()
    ):

        row = {

            'chart_type':
                chart_type,

            'question_type':
                question_type,

            # ------------------------------------------------
            # Total sampel
            # ------------------------------------------------

            'count':
                metrics_dict['count'],

            # ------------------------------------------------
            # Jumlah sampel yang diukur
            # oleh masing-masing metrik
            # ------------------------------------------------

            'em_count':
                metrics_dict['em_count'],

            'nls_without_threshold_count':
                metrics_dict[
                    'nls_without_threshold_count'
                ],

            'nls_with_threshold_count':
                metrics_dict[
                    'nls_with_threshold_count'
                ],

            'relaxed_acc_count':
                metrics_dict[
                    'relaxed_acc_count'
                ],

            # ------------------------------------------------
            # Overall score
            #
            # Text -> NLS without threshold
            # Year -> EM
            # Numeric -> Relaxed Accuracy
            # ------------------------------------------------

            'avg_score':
                np.mean(
                    metrics_dict['scores']
                ),

            # ------------------------------------------------
            # Numeric
            # ------------------------------------------------

            'avg_relaxed_acc':
                np.mean(
                    metrics_dict[
                        'relaxed_acc_scores'
                    ]
                )
                if metrics_dict[
                    'relaxed_acc_scores'
                ]
                else None,

            # ------------------------------------------------
            # Text + Year
            # ------------------------------------------------

            'avg_em':
                np.mean(
                    metrics_dict[
                        'em_scores'
                    ]
                )
                if metrics_dict[
                    'em_scores'
                ]
                else None,

            'avg_nls_without_threshold':
                np.mean(
                    metrics_dict[
                        'nls_without_threshold_scores'
                    ]
                )
                if metrics_dict[
                    'nls_without_threshold_scores'
                ]
                else None,

            'avg_nls_with_threshold':
                np.mean(
                    metrics_dict[
                        'nls_with_threshold_scores'
                    ]
                )
                if metrics_dict[
                    'nls_with_threshold_scores'
                ]
                else None,
        }

        data.append(row)

    df = pd.DataFrame(
        data
    )

    # ========================================================
    # ROUND
    # ========================================================

    numeric_cols = [
        'avg_score',
        'avg_relaxed_acc',
        'avg_em',
        'avg_nls_without_threshold',
        'avg_nls_with_threshold'
    ]

    for col in numeric_cols:

        if col in df.columns:

            df[col] = df[col].round(4)

    # ========================================================
    # SUMMARY
    # ========================================================

    print(
        f"\nTotal records processed: "
        f"{total_records}"
    )

    print(
        f"Resolved + evaluated JSONL saved to: "
        f"{resolved_directory}"
    )

    # ========================================================
    # RANKING TYPE SUMMARY
    # ========================================================

    if ranking_answer_types:

        print(
            "\nRanking answer types:"
        )

        for (
            answer_type,
            count
        ) in ranking_answer_types.items():

            print(
                f"  {answer_type}: "
                f"{count}"
            )

    # ========================================================
    # AVERAGE TYPE SUMMARY
    # ========================================================

    if average_answer_types:

        print(
            "\nAverage answer types:"
        )

        for (
            answer_type,
            count
        ) in average_answer_types.items():

            print(
                f"  {answer_type}: "
                f"{count}"
            )

    # ========================================================
    # OVERALL METRICS
    # ========================================================

    all_scores = []
    all_em_scores = []
    all_nls_without_threshold_scores = []
    all_nls_with_threshold_scores = []
    all_relaxed_acc_scores = []

    for group_metrics in results.values():

        all_scores.extend(
            group_metrics['scores']
        )

        all_em_scores.extend(
            group_metrics['em_scores']
        )

        all_nls_without_threshold_scores.extend(
            group_metrics[
                'nls_without_threshold_scores'
            ]
        )

        all_nls_with_threshold_scores.extend(
            group_metrics[
                'nls_with_threshold_scores'
            ]
        )

        all_relaxed_acc_scores.extend(
            group_metrics[
                'relaxed_acc_scores'
            ]
        )

    # --------------------------------------------------------
    # Overall Average Score
    # Semua sample
    # --------------------------------------------------------

    overall_avg_score = (
        np.mean(all_scores)
        if all_scores
        else None
    )

    # --------------------------------------------------------
    # Overall Average Relaxed Accuracy
    # Numeric saja
    # --------------------------------------------------------

    overall_avg_relaxed_acc = (
        np.mean(all_relaxed_acc_scores)
        if all_relaxed_acc_scores
        else None
    )

    # --------------------------------------------------------
    # Overall Average EM
    # Text + Year
    # --------------------------------------------------------

    overall_avg_em = (
        np.mean(all_em_scores)
        if all_em_scores
        else None
    )

    # --------------------------------------------------------
    # Overall Average NLS without threshold
    # Text + Year
    # --------------------------------------------------------

    overall_avg_nls_without_threshold = (
        np.mean(
            all_nls_without_threshold_scores
        )
        if all_nls_without_threshold_scores
        else None
    )

    # --------------------------------------------------------
    # Overall Average NLS with threshold
    # Text + Year
    # --------------------------------------------------------

    overall_avg_nls_with_threshold = (
        np.mean(
            all_nls_with_threshold_scores
        )
        if all_nls_with_threshold_scores
        else None
    )

    # ========================================================
    # JUMLAH SAMPEL UNTUK OVERALL METRIK
    # ========================================================

    overall_em_count = len(
        all_em_scores
    )

    overall_nls_without_threshold_count = len(
        all_nls_without_threshold_scores
    )

    overall_nls_with_threshold_count = len(
        all_nls_with_threshold_scores
    )

    overall_relaxed_acc_count = len(
        all_relaxed_acc_scores
    )

    # ========================================================
    # RETURN
    # ========================================================

    summary = {

        'total_samples':
            total_records,

        'overall_avg_score':
            overall_avg_score,

        'overall_avg_relaxed_acc':
            overall_avg_relaxed_acc,

        'overall_avg_em':
            overall_avg_em,

        'overall_avg_nls_without_threshold':
            overall_avg_nls_without_threshold,

        'overall_avg_nls_with_threshold':
            overall_avg_nls_with_threshold,

        # ----------------------------------------------------
        # Jumlah sampel tiap metrik
        # ----------------------------------------------------

        'overall_em_count':
            overall_em_count,

        'overall_nls_without_threshold_count':
            overall_nls_without_threshold_count,

        'overall_nls_with_threshold_count':
            overall_nls_with_threshold_count,

        'overall_relaxed_acc_count':
            overall_relaxed_acc_count
    }

    return df, summary


# ============================================================
# MAIN
# ============================================================

if __name__ == '__main__':

    # ========================================================
    # INPUT DIRECTORY
    # ========================================================

    # jsonl_directory = (
    #     "/kaggle/input/datasets/"
    #     "dianshinta09/eval-results-ver2/"
    # )

    jsonl_directory = (
        "/kaggle/input/datasets/"
        "dianshinta09/evaluation-result-tinycart/eval/"
    )

    # ========================================================
    # EVALUATION
    # ========================================================

    evaluation_output = evaluate_jsonl_files(
        jsonl_directory
    )

    if evaluation_output is not None:

        df_results, summary = evaluation_output

        print(
            "\n"
            + "=" * 80
        )

        print(
            "EVALUATION RESULTS"
        )

        print(
            "=" * 80
        )

        print(
            df_results.to_string(
                index=False
            )
        )

        print(
            "=" * 80
        )

        # ====================================================
        # SUMMARY
        # ====================================================

        print(
            "\nSUMMARY:"
        )

        print(
            f"Total chart types: "
            f"{df_results['chart_type'].nunique()}"
        )

        print(
            f"Total question types: "
            f"{df_results['question_type'].nunique()}"
        )

        print(
            f"Total samples: "
            f"{summary['total_samples']}"
        )

        print(
            f"Overall Average Score: "
            f"{summary['overall_avg_score']:.4f}"
        )

        print(
            f"Overall Average Exact Match "
            f"(Text + Year): "
            f"{summary['overall_avg_em']:.4f}"
            f" "
            f"(n={summary['overall_em_count']})"
        )

        print(
            f"Overall Average Normalized Lev Similarity "
            f"without threshold (Text + Year): "
            f"{summary['overall_avg_nls_without_threshold']:.4f}"
            f" "
            f"(n={summary['overall_nls_without_threshold_count']})"
        )

        print(
            f"Overall Average Normalized Lev Similarity "
            f"with threshold (Text + Year): "
            f"{summary['overall_avg_nls_with_threshold']:.4f}"
            f" "
            f"(n={summary['overall_nls_with_threshold_count']})"
        )

        print(
            f"Overall Average Relaxed Accuracy "
            f"(Numeric): "
            f"{summary['overall_avg_relaxed_acc']:.4f}"
            f" "
            f"(n={summary['overall_relaxed_acc_count']})"
        )

        # ====================================================
        # SAVE CSV
        # ====================================================

        output_file = (
            '/kaggle/working/'
            'evaluation_results.csv'
        )

        df_results.to_csv(
            output_file,
            index=False
        )

        print(
            f"\n✓ Results saved to "
            f"{output_file}"
        )

        # ====================================================
        # ZIP RESOLVED + EVALUATED JSONL
        # ====================================================

        resolved_directory = Path(
            '/kaggle/working/resolved_jsonl'
        )

        zip_file = (
            '/kaggle/working/'
            'resolved_jsonl.zip'
        )

        jsonl_files_resolved = sorted(
            resolved_directory.glob(
                '*.jsonl'
            )
        )

        with zipfile.ZipFile(
            zip_file,
            'w',
            zipfile.ZIP_DEFLATED
        ) as zipf:

            for jsonl_file in jsonl_files_resolved:

                zipf.write(
                    jsonl_file,
                    arcname=jsonl_file.name
                )

        # ====================================================
        # ZIP INFO
        # ====================================================

        zip_size_mb = (
            os.path.getsize(
                zip_file
            )
            / (1024 * 1024)
        )

        print(
            f"\n✓ Resolved + evaluated JSONL ZIP saved to: "
            f"{zip_file}"
        )

        print(
            f"✓ Total JSONL files in ZIP: "
            f"{len(jsonl_files_resolved)}"
        )

        print(
            f"✓ ZIP size: "
            f"{zip_size_mb:.2f} MB"
        )
```

```python
def extract_qa(row):
    """
    Mengambil question dan answer dari struktur conversations.
    """

    question = ""
    answer = ""

    # Format conversations
    if "conversations" in row and row["conversations"] is not None:
        conv = row["conversations"]

        for item in conv:
            role = item.get("from", "")
            value = item.get("value", "")

            if role == "human":
                question = value

            elif role == "gpt":
                answer = value

    # Kalau dataset langsung punya kolom question/answer
    elif "question" in row:
        question = row.get("question", "")

        if "answer" in row:
            answer = row.get("answer", "")
        elif "gt_answer" in row:
            answer = row.get("gt_answer", "")

    return question, answer

def get_id(row):
    return str(row.get("id", ""))


def detect_question_type(row):
    id_value = get_id(row)

    # Prioritas dari ID
    patterns = [
        ("Selection", r"Selection"),
        ("Average", r"Average"),
        ("Ranking", r"Ranking"),
        ("Compare", r"Compare"),
        ("Min", r"Min"),
        ("Max", r"Max"),
        ("Sum", r"Sum"),
    ]

    for qtype, pattern in patterns:
        if re.search(pattern, id_value, flags=re.IGNORECASE):
            return qtype

    return "Unknown"

def detect_chart_type(row):
    id_value = get_id(row).lower()

    if "gbar" in id_value:
        return "GBar"

    if "bar" in id_value:
        return "Bar"

    if "line" in id_value:
        return "Line"

    return "Unknown"

def detect_pot(question):
    if "jawab menggunakan langkah perhitungannya" in question.lower():
        return "PoT"

    return "Non-PoT"

def dataset_to_df(ds):

    records = []

    for row in ds:

        question, answer = extract_qa(row)

        records.append({
            "id": get_id(row),
            "question": question,
            "answer": answer,
            "question_type": detect_question_type(row),
            "chart_type": detect_chart_type(row),
            "pot_type": detect_pot(question),
            "question_length": len(question),
            "answer_length": len(str(answer)),
        })

    return pd.DataFrame(records)


df1 = dataset_to_df(ds1)
df2 = dataset_to_df(ds2)

summary = pd.DataFrame({
    "Dataset 1": [
        len(df1),
        df1["question_type"].nunique(),
        df1["chart_type"].nunique(),
    ],
    "Dataset 2": [
        len(df2),
        df2["question_type"].nunique(),
        df2["chart_type"].nunique(),
    ]
}, index=[
    "Jumlah train",
    "Jumlah question type",
    "Jumlah chart type"
])

display(summary)

question_dist = pd.concat([
    df1["question_type"].value_counts().rename("Dataset 1"),
    df2["question_type"].value_counts().rename("Dataset 2")
], axis=1).fillna(0)

question_dist["Dataset 1 (%)"] = (
    question_dist["Dataset 1"] / len(df1) * 100
)

question_dist["Dataset 2 (%)"] = (
    question_dist["Dataset 2"] / len(df2) * 100
)

question_dist = question_dist.sort_index()

display(question_dist)

chart_dist = pd.concat([
    df1["chart_type"].value_counts().rename("Dataset 1"),
    df2["chart_type"].value_counts().rename("Dataset 2")
], axis=1).fillna(0)

chart_dist["Dataset 1 (%)"] = (
    chart_dist["Dataset 1"] / len(df1) * 100
)

chart_dist["Dataset 2 (%)"] = (
    chart_dist["Dataset 2"] / len(df2) * 100
)

chart_dist = chart_dist.sort_index()

display(chart_dist)
```

```python
# ============================================================
# TARGET:
#   - 100 PoT
#   - 400 non-PoT
#
# STRATA:
#   Region × Chart Type × Question Type
#
# ATURAN:
#   - PoT dan non-PoT disampling secara stratified
#
# OUTPUT:
# 1. CSV berisi 500 sample + score
# 2. Folder berisi 500 gambar chart
#
# RANDOM SEED = 42
# ============================================================

!pip install -q datasets

import random
import pandas as pd
import json
import glob
import os
import re
import math

from collections import defaultdict
from datasets import load_dataset


# ============================================================
# CONFIG
# ============================================================

DATASET_NAME = "shiinn97/indochartqa"

SAMPLE_SIZE = 500
POT_SIZE = 100
NONPOT_SIZE = 400

RANDOM_SEED = 42

POT_PHRASE = "Jawab menggunakan langkah perhitungannya"

EVAL_DIR = "/kaggle/input/datasets/dianshinta09/evaluation-result-tinycart/eval"
IMAGE_DIR = "/kaggle/working/tinychart_images/tinychart_images/"
OUTPUT_FILE = "/kaggle/working/indochartqa_test_sample_500.csv"

os.makedirs(IMAGE_DIR, exist_ok=True)

random.seed(RANDOM_SEED)


# ============================================================
# FUNGSI PARSE ID
# ============================================================

pattern = re.compile(
    r"^([BS])\d+(BAR|GBAR|LINE)"
    r"(Sum|Average|Min|Max|Ranking|Compare|Selection)_\d+$"
)


def parse_id(sample_id):

    match = pattern.match(str(sample_id))

    if not match:
        return None

    region = match.group(1)
    chart_type = match.group(2)
    question_type = match.group(3)

    return (
        region,
        chart_type,
        question_type
    )


# ============================================================
# LOAD SCORE
# ============================================================

print("=" * 70)
print("MEMBACA SCORE")
print("=" * 70)

json_files = glob.glob(
    os.path.join(EVAL_DIR, "*.json")
)

jsonl_files = glob.glob(
    os.path.join(EVAL_DIR, "*.jsonl")
)

eval_files = json_files + jsonl_files

print(
    "Jumlah file evaluasi:",
    len(eval_files)
)

score_dict = {}

for file_path in eval_files:

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            data = json.loads(line)

            if (
                "id" in data
                and "score" in data
            ):

                score_dict[
                    str(data["id"])
                ] = data["score"]


print(
    "Jumlah ID yang memiliki score:",
    len(score_dict)
)


# ============================================================
# LOAD TEST SET
# ============================================================

print("\n" + "=" * 70)
print("LOAD TEST SET")
print("=" * 70)

dataset = load_dataset(DATASET_NAME)

test_data = dataset["test"]

print(
    "Jumlah seluruh test set:",
    len(test_data)
)


# ============================================================
# FUNGSI AMBIL HUMAN QUESTION
# ============================================================

def get_human_question(conversations):

    if isinstance(conversations, str):

        conversations = json.loads(
            conversations
        )

    for message in conversations:

        if message.get("from") == "human":

            question = message.get(
                "value",
                ""
            )

            question = (
                question
                .replace("<image>\n", "")
                .replace("<image>", "")
                .strip()
            )

            return question

    return ""


# ============================================================
# FUNGSI CEK PoT
# ============================================================

def is_pot(sample):

    question = get_human_question(
        sample["conversations"]
    )

    return (
        POT_PHRASE.lower()
        in question.lower()
    )


# ============================================================
# FILTER:
# HANYA DATA YANG MEMILIKI SCORE
# ============================================================

eligible_indices = []

for idx, sample in enumerate(test_data):

    sample_id = str(sample["id"])

    if sample_id in score_dict:

        # Pastikan ID dapat diparse
        if parse_id(sample_id) is not None:

            eligible_indices.append(idx)


print("\n" + "=" * 70)
print("ELIGIBLE DATA")
print("=" * 70)

print(
    "Data yang memiliki score:",
    len(eligible_indices)
)


# ============================================================
# KELOMPOKKAN PoT DAN NON-PoT
# ============================================================

pot_indices = []
nonpot_indices = []

for idx in eligible_indices:

    sample = test_data[idx]

    if is_pot(sample):

        pot_indices.append(idx)

    else:

        nonpot_indices.append(idx)


print("\n" + "=" * 70)
print("DISTRIBUSI PoT / NON-PoT")
print("=" * 70)

print(
    "PoT     :",
    len(pot_indices)
)

print(
    "non-PoT :",
    len(nonpot_indices)
)


# ============================================================
# KELOMPOKKAN BERDASARKAN STRATA
#
# Region × Chart Type × Question Type
# ============================================================

pot_strata = defaultdict(list)

nonpot_strata = defaultdict(list)


for idx in pot_indices:

    parsed = parse_id(
        str(test_data[idx]["id"])
    )

    region, chart_type, question_type = parsed

    key = (
        region,
        chart_type,
        question_type
    )

    pot_strata[key].append(idx)


for idx in nonpot_indices:

    parsed = parse_id(
        str(test_data[idx]["id"])
    )

    region, chart_type, question_type = parsed

    key = (
        region,
        chart_type,
        question_type
    )

    nonpot_strata[key].append(idx)


# ============================================================
# FUNGSI ALOKASI PROPORSIONAL
# ============================================================

def proportional_allocation(
    strata,
    target_size
):

    total_population = sum(
        len(indices)
        for indices in strata.values()
    )

    allocation = {}

    for key, indices in strata.items():

        population = len(indices)

        ideal = (
            population
            / total_population
            * target_size
        )

        allocation[key] = {
            "population": population,
            "ideal": ideal,
            "sample": math.floor(ideal),
            "remainder":
                ideal - math.floor(ideal)
        }


    # --------------------------------------------------------
    # Tambahkan sisa berdasarkan remainder terbesar
    # --------------------------------------------------------

    current_total = sum(
        x["sample"]
        for x in allocation.values()
    )

    difference = (
        target_size
        - current_total
    )

    candidates = sorted(
        allocation.keys(),
        key=lambda x:
            allocation[x]["remainder"],
        reverse=True
    )

    while difference > 0:

        changed = False

        for key in candidates:

            if difference == 0:
                break

            if (
                allocation[key]["sample"]
                <
                allocation[key]["population"]
            ):

                allocation[key]["sample"] += 1

                difference -= 1

                changed = True

        if not changed:
            break


    return allocation


# ============================================================
# ALOKASI PoT
# ============================================================

pot_allocation = proportional_allocation(
    pot_strata,
    POT_SIZE
)


# ============================================================
# ALOKASI NON-PoT
# ============================================================

nonpot_allocation = proportional_allocation(
    nonpot_strata,
    NONPOT_SIZE
)


# ============================================================
# VALIDASI ALOKASI
# ============================================================

pot_allocated = sum(
    x["sample"]
    for x in pot_allocation.values()
)

nonpot_allocated = sum(
    x["sample"]
    for x in nonpot_allocation.values()
)

print("\n" + "=" * 70)
print("VALIDASI ALOKASI")
print("=" * 70)

print(
    "PoT allocated     :",
    pot_allocated
)

print(
    "non-PoT allocated :",
    nonpot_allocated
)

print(
    "Total             :",
    pot_allocated
    + nonpot_allocated
)

assert pot_allocated == POT_SIZE

assert nonpot_allocated == NONPOT_SIZE


# ============================================================
# RANDOM SAMPLING DALAM SETIAP STRATA
# ============================================================

selected_pot = []

selected_nonpot = []

summary = []


# ------------------------------------------------------------
# PoT
# ------------------------------------------------------------

for key in sorted(pot_strata.keys()):

    region, chart_type, question_type = key

    indices = pot_strata[key]

    n_sample = pot_allocation[key]["sample"]

    selected = random.sample(
        indices,
        n_sample
    )

    selected_pot.extend(
        selected
    )


# ------------------------------------------------------------
# NON-PoT
# ------------------------------------------------------------

for key in sorted(nonpot_strata.keys()):

    region, chart_type, question_type = key

    indices = nonpot_strata[key]

    n_sample = nonpot_allocation[key]["sample"]

    selected = random.sample(
        indices,
        n_sample
    )

    selected_nonpot.extend(
        selected
    )


# ============================================================
# GABUNGKAN DAN ACAK
# ============================================================

selected_indices = (
    selected_pot
    +
    selected_nonpot
)

random.shuffle(
    selected_indices
)


# ============================================================
# AMBIL DATA
# ============================================================

sampled_dataset = test_data.select(
    selected_indices
)

df = sampled_dataset.to_pandas()


# ============================================================
# TAMBAHKAN SCORE
# ============================================================

df["score"] = (
    df["id"]
    .astype(str)
    .map(score_dict)
)


# ============================================================
# TAMBAHKAN LABEL PoT / NON-PoT
# ============================================================

df["answer_type"] = df["conversations"].apply(
    lambda conv:
        "PoT"
        if is_pot({
            "conversations": conv
        })
        else "non-PoT"
)


# ============================================================
# TAMBAHKAN INFORMASI STRATA
# ============================================================

parsed_info = (
    df["id"]
    .astype(str)
    .apply(parse_id)
)

df["region"] = parsed_info.apply(
    lambda x: x[0]
)

df["chart_type"] = parsed_info.apply(
    lambda x: x[1]
)

df["question_type"] = parsed_info.apply(
    lambda x: x[2]
)


# ============================================================
# VALIDASI FINAL
# ============================================================

print("\n" + "=" * 70)
print("VALIDASI SAMPLE")
print("=" * 70)

print(
    "Jumlah sample:",
    len(df)
)

print(
    "PoT:",
    (df["answer_type"] == "PoT").sum()
)

print(
    "non-PoT:",
    (df["answer_type"] == "non-PoT").sum()
)

print(
    "Score tersedia:",
    df["score"].notna().sum()
)

print(
    "Score missing:",
    df["score"].isna().sum()
)

print(
    "Selection:",
    (df["question_type"] == "Selection").sum()
)


assert len(df) == SAMPLE_SIZE

assert (
    (df["answer_type"] == "PoT").sum()
    == POT_SIZE
)

assert (
    (df["answer_type"] == "non-PoT").sum()
    == NONPOT_SIZE
)

assert df["score"].notna().all()


# ============================================================
# AVERAGE SCORE
# ============================================================

average_score = df["score"].mean()

print(
    f"\nAverage Score: "
    f"{average_score:.4f}"
)

print(
    f"Average Score (%): "
    f"{average_score * 100:.2f}%"
)


# ============================================================
# DISTRIBUSI PoT / NON-PoT
# ============================================================

distribution = (
    df["answer_type"]
    .value_counts()
    .rename_axis("Answer_Type")
    .reset_index(name="Count")
)

distribution["Percentage"] = (
    distribution["Count"]
    / len(df)
    * 100
)


print("\n" + "=" * 70)
print("DISTRIBUSI PoT / NON-PoT")
print("=" * 70)

display(distribution)


# ============================================================
# DISTRIBUSI STRATA FINAL
# ============================================================

final_distribution = (
    df.groupby(
        [
            "region",
            "chart_type",
            "question_type",
            "answer_type"
        ]
    )
    .size()
    .reset_index(name="Count")
)


print("\n" + "=" * 70)
print("DISTRIBUSI FINAL PER STRATA")
print("=" * 70)

display(
    final_distribution
)


# ============================================================
# SIMPAN CSV
# ============================================================

df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ============================================================
# SELESAI
# ============================================================

print("\n" + "=" * 70)
print("STRATIFIED SAMPLING SELESAI")
print("=" * 70)

print(
    "Test set awal :",
    len(test_data)
)

print(
    "Eligible      :",
    len(eligible_indices)
)

print(
    "Jumlah strata PoT:",
    len(pot_strata)
)

print(
    "Jumlah strata non-PoT:",
    len(nonpot_strata)
)

print(
    "Sample        :",
    len(df)
)

print(
    "PoT sample    :",
    POT_SIZE
)

print(
    "non-PoT sample:",
    NONPOT_SIZE
)

print(
    f"Average Score : "
    f"{average_score:.4f} "
    f"({average_score * 100:.2f}%)"
)

print(
    "\nCSV   :",
    OUTPUT_FILE
)
```

```python
import os
import ast
import base64
import pandas as pd
from tqdm import tqdm
from openai import OpenAI
import time
import random

# ============================================================
# PATH
# ============================================================

CSV_PATH = "/kaggle/working/indochartqa_test_sample_500.csv"
IMAGE_DIR = "/kaggle/working/indochartqa_test_sample_500_images/"

OUTPUT_PATH = "/kaggle/working/gpt4o_mini_results_500.csv"

MODEL = "gpt-4o-mini"

# ============================================================
# SETTINGS
# ============================================================

# Jeda antar-request agar tidak terlalu cepat mencapai TPM limit
REQUEST_DELAY = 3

# Jumlah maksimal percobaan jika terkena rate limit
MAX_RETRIES = 6

# Simpan hasil setiap N data
SAVE_EVERY = 10

# ============================================================
# OPENAI
# ============================================================

client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])

# ============================================================
# LOAD CSV
# ============================================================

df = pd.read_csv(CSV_PATH)

print("Jumlah data:", len(df))

CONVERSATION_COLUMN = "conversations"

print("Conversation column:", CONVERSATION_COLUMN)


# ============================================================
# PARSE HUMAN DAN GPT
# ============================================================

def extract_human_question(text):
    """
    Mengambil value dari:
    {'from': 'human', 'value': '...'}

    Hasilnya hanya question, tanpa <image>.
    """

    text = str(text)

    marker_from = "'from': 'human'"
    marker_value = "'value':"

    start = text.find(marker_from)

    if start == -1:
        return None

    value_start = text.find(marker_value, start)

    if value_start == -1:
        return None

    value_start += len(marker_value)

    while value_start < len(text) and text[value_start].isspace():
        value_start += 1

    if text[value_start] == "'":
        value_start += 1

        value_end = value_start

        while value_end < len(text):
            if text[value_end] == "'" and text[value_end - 1] != "\\":
                break
            value_end += 1

        question = text[value_start:value_end]

    else:
        value_end = text.find("'}", value_start)

        if value_end == -1:
            value_end = len(text)

        question = text[value_start:value_end]

    question = question.replace("<image>", "").strip()

    return question


def extract_gpt_answer(text):
    """
    Mengambil value dari:
    {'from': 'gpt', 'value': '...'}

    Hanya digunakan untuk menyimpan ground truth.
    Tidak dikirim ke GPT-4o-mini.
    """

    text = str(text)

    marker_from = "'from': 'gpt'"
    marker_value = "'value':"

    start = text.find(marker_from)

    if start == -1:
        return None

    value_start = text.find(marker_value, start)

    if value_start == -1:
        return None

    value_start += len(marker_value)

    while value_start < len(text) and text[value_start].isspace():
        value_start += 1

    if text[value_start] == "'":
        value_start += 1

        value_end = value_start

        while value_end < len(text):
            if text[value_end] == "'" and text[value_end - 1] != "\\":
                break
            value_end += 1

        answer = text[value_start:value_end]

    else:
        value_end = text.find("'}", value_start)

        if value_end == -1:
            value_end = len(text)

        answer = text[value_start:value_end]

    return answer


# ============================================================
# IMAGE PATH
# ============================================================

def find_image(image_id):

    image_name = f"{image_id}.png"
    image_path = os.path.join(IMAGE_DIR, image_name)

    if os.path.exists(image_path):
        return image_path

    return None


# ============================================================
# IMAGE -> BASE64
# ============================================================

def image_to_data_url(image_path):

    ext = os.path.splitext(image_path)[1].lower()

    mime_types = {
        ".jpg": "image/jpeg",
        ".jpeg": "image/jpeg",
        ".png": "image/png",
    }

    if ext not in mime_types:
        raise ValueError(f"Format image tidak didukung: {ext}")

    with open(image_path, "rb") as f:
        encoded = base64.b64encode(f.read()).decode("utf-8")

    return f"data:{mime_types[ext]};base64,{encoded}"


# ============================================================
# CEK NAMA KOLOM IMAGE
# ============================================================

print("\nPreview data:")
print(df.head(1).to_string())

IMAGE_COLUMN = "id"

if IMAGE_COLUMN not in df.columns:

    print("\nWARNING:")
    print(f"Kolom '{IMAGE_COLUMN}' tidak ditemukan.")
    print("Kolom yang tersedia:", df.columns.tolist())

    raise ValueError(
        "Silakan ubah IMAGE_COLUMN sesuai nama kolom image di CSV."
    )


# ============================================================
# SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """
Aturan output:

1. Jika pertanyaan TIDAK mengandung instruksi
"Jawab menggunakan langkah perhitungannya.",
jawab hanya dengan jawaban akhir yang singkat dan tepat.

Jangan memberikan:
- penjelasan
- alasan
- langkah perhitungan
- narasi tambahan
- satuan

2. Jika pertanyaan MENGANDUNG instruksi
"Jawab menggunakan langkah perhitungannya.",
jawaban HARUS berupa Program of Thoughts (PoT)
dalam bentuk assignment statement Python.

Jangan memberikan narasi atau penjelasan di luar format PoT.

Untuk jawaban PoT:

- Gunakan informasi pada gambar grafik untuk menentukan nilai
  yang diperlukan dalam perhitungan.
- Gunakan komentar satu baris sebelum setiap assignment statement.
- Setiap komentar dibungkus dengan tag <comment>#...</comment>.
- Setiap assignment statement dibungkus dengan tag <step>...</step>.
- Komentar harus bersifat generik dan deskriptif secara logika.
- Output akhir harus disimpan dalam variabel Answer.
- Jangan menggunakan Markdown code fence.
- Jangan memberikan narasi di luar format PoT.

Hanya boleh menggunakan operator dan fungsi berikut:

['len', 'all', 'any', 'index', 'np.sort', 'np.abs',
'np.add', 'np.argmax', 'np.argmin', 'np.diff', 'np.divide',
'np.greater', 'np.greater_equal', 'np.less', 'np.max',
'np.mean', 'np.median', 'np.min', 'np.subtract', 'np.sum',
'np.count_nonzero', 'np.where', '+', '-', '*', '/', '>',
'<', '=']
"""


# ============================================================
# FUNCTION: REQUEST GPT
# ============================================================

def request_gpt(question, image_data_url):

    for attempt in range(MAX_RETRIES):

        try:

            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "system",
                        "content": SYSTEM_PROMPT
                    },
                    {
                        "role": "user",
                        "content": [
                            {
                                "type": "text",
                                "text": question
                            },
                            {
                                "type": "image_url",
                                "image_url": {
                                    "url": image_data_url
                                }
                            }
                        ]
                    }
                ],
                temperature=0,
                max_tokens=500
            )

            return response.choices[0].message.content.strip(), None

        except Exception as e:

            error_message = str(e)

            # -----------------------------------------------
            # RATE LIMIT
            # -----------------------------------------------

            if "429" in error_message:

                # Exponential backoff:
                # 5 -> 10 -> 20 -> 40 -> 60 -> 60 detik
                wait_time = min(
                    60,
                    5 * (2 ** attempt)
                )

                # Tambahkan sedikit random delay
                wait_time += random.uniform(0, 2)

                print(
                    f"\n[429] Rate limit. "
                    f"Percobaan {attempt + 1}/{MAX_RETRIES}. "
                    f"Tunggu {wait_time:.1f} detik..."
                )

                time.sleep(wait_time)

            else:

                return None, error_message

    return None, "rate_limit_after_retries"


# ============================================================
# RUN GPT-4o-mini
# ============================================================

results = []

for idx, row in tqdm(
    df.iterrows(),
    total=len(df),
    desc="Running GPT-4o-mini"
):

    conversation = row[CONVERSATION_COLUMN]

    # --------------------------------------------------------
    # QUESTION
    # --------------------------------------------------------

    question = extract_human_question(conversation)

    # --------------------------------------------------------
    # GROUND TRUTH
    # --------------------------------------------------------

    ground_truth = extract_gpt_answer(conversation)

    # --------------------------------------------------------
    # IMAGE
    # --------------------------------------------------------

    image_id = str(row[IMAGE_COLUMN])

    image_path = find_image(image_id)

    # --------------------------------------------------------
    # QUESTION TIDAK DITEMUKAN
    # --------------------------------------------------------

    if question is None:

        results.append({
            "index": idx,
            "image": image_id,
            "question": None,
            "ground_truth": ground_truth,
            "gpt4o_mini_answer": None,
            "error": "question_not_found"
        })

        continue

    # --------------------------------------------------------
    # IMAGE TIDAK DITEMUKAN
    # --------------------------------------------------------

    if image_path is None:

        results.append({
            "index": idx,
            "image": image_id,
            "question": question,
            "ground_truth": ground_truth,
            "gpt4o_mini_answer": None,
            "error": "image_not_found"
        })

        continue

    # --------------------------------------------------------
    # IMAGE -> DATA URL
    # --------------------------------------------------------

    try:

        image_data_url = image_to_data_url(image_path)

    except Exception as e:

        results.append({
            "index": idx,
            "image": image_id,
            "question": question,
            "ground_truth": ground_truth,
            "gpt4o_mini_answer": None,
            "error": str(e)
        })

        continue

    # --------------------------------------------------------
    # DELAY ANTAR REQUEST
    # --------------------------------------------------------

    time.sleep(REQUEST_DELAY)

    # --------------------------------------------------------
    # REQUEST GPT
    # --------------------------------------------------------

    answer, error = request_gpt(
        question,
        image_data_url
    )

    results.append({
        "index": idx,
        "image": image_id,
        "question": question,
        "ground_truth": ground_truth,
        "gpt4o_mini_answer": answer,
        "error": error
    })

    # --------------------------------------------------------
    # AUTO SAVE
    # --------------------------------------------------------

    if len(results) % SAVE_EVERY == 0:

        temp_df = pd.DataFrame(results)

        temp_df.to_csv(
            OUTPUT_PATH,
            index=False,
            encoding="utf-8-sig"
        )


# ============================================================
# FINAL SAVE
# ============================================================

result_df = pd.DataFrame(results)

result_df.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# SUMMARY
# ============================================================

print("\n========================================")
print("SELESAI")
print("========================================")

print("Output :", OUTPUT_PATH)
print("Total  :", len(result_df))

print(
    "Berhasil:",
    result_df["gpt4o_mini_answer"].notna().sum()
)

print(
    "Gagal:",
    result_df["gpt4o_mini_answer"].isna().sum()
)
```

```python
import json
import re
import os
from datasets import load_dataset
from pathlib import Path
from collections import defaultdict
from difflib import SequenceMatcher
from openpyxl import load_workbook
from openpyxl.drawing.image import Image as XLImage
from openpyxl.utils import get_column_letter
from PIL import Image
import tempfile
import pandas as pd
import numpy as np


# ============================================================
# PATH
# ============================================================

CSV_PATH = "/kaggle/input/datasets/dianshinta09/mygptresult/gpt4o_mini_results_500.csv"

# Folder JSONL TinyChart evaluation
JSONL_DIR = "/kaggle/input/datasets/dianshinta09/evaluation-result-tinycart/eval"
OUTPUT_PATH = "/kaggle/working/evaluation_gpt4o_mini_500.xlsx"
HF_DATASET = "shiinn97/indochartqa"
HF_SPLIT = "test"

IMAGE_ROOT = "/kaggle/working/tinychart_images/tinychart_images/"

# ============================================================
# KONFIGURASI
# ============================================================

POT_SUFFIX = "Jawab menggunakan langkah perhitungannya"


# ============================================================
# 1. LOAD JSONL UNTUK GT PoT
# ============================================================

def load_pot_ground_truth(jsonl_directory):
    """
    Membaca seluruh JSONL dan membuat mapping:
    
        id -> {
            question,
            final_gt_answer,
            ...
        }
    
    Hanya digunakan untuk mengambil final_gt_answer
    pada sampel PoT.
    """

    pot_gt = {}

    jsonl_files = list(Path(jsonl_directory).glob("*.jsonl"))

    print(f"Found {len(jsonl_files)} JSONL files")

    for jsonl_file in sorted(jsonl_files):

        try:
            with open(jsonl_file, "r", encoding="utf-8") as f:

                for line in f:

                    if not line.strip():
                        continue

                    record = json.loads(line)

                    record_id = record.get("id")

                    if record_id is None:
                        continue

                    pot_gt[record_id] = {
                        "question": record.get("question", ""),
                        "final_gt_answer": record.get(
                            "final_gt_answer", ""
                        ),
                        "gt_answer": record.get(
                            "gt_answer", ""
                        )
                    }

        except Exception as e:
            print(f"ERROR reading {jsonl_file.name}: {e}")

    print(f"Loaded {len(pot_gt)} JSONL records")

    return pot_gt

def load_hf_image_mapping():

    print("=" * 80)
    print("Loading Hugging Face test set")
    print("=" * 80)

    dataset = load_dataset(
        HF_DATASET,
        split=HF_SPLIT
    )

    print(f"HF test records : {len(dataset)}")
    print(f"Columns         : {dataset.column_names}")

    image_mapping = {}

    for row in dataset:

        image_id = row.get("id")

        if image_id is None:
            continue

        image_id = os.path.splitext(
            os.path.basename(str(image_id).strip())
        )[0]

        # ----------------------------------------------------
        # Ambil path dari kolom image
        # ----------------------------------------------------

        image_value = row.get("image")

        image_path = None

        if isinstance(image_value, str):

            image_path = image_value

        elif isinstance(image_value, dict):

            # Beberapa dataset HF menyimpan image sebagai:
            # {"path": "...", "bytes": ...}

            image_path = image_value.get("path")

        # ----------------------------------------------------
        # Simpan mapping
        # ----------------------------------------------------

        if image_path:

            image_mapping[image_id] = image_path

    print(
        f"Image path mapping : {len(image_mapping)}"
    )

    return image_mapping

def resolve_local_image_path(image_id, image_mapping):

    image_id = os.path.splitext(
        os.path.basename(str(image_id).strip())
    )[0]

    hf_path = image_mapping.get(image_id)

    if hf_path is None:
        return None

    # Path dari Hugging Face adalah path relatif
    hf_path = str(hf_path).replace("\\", "/").lstrip("/")

    # Join dengan root folder lokal
    local_path = os.path.join(
        IMAGE_ROOT,
        hf_path
    )

    return local_path
    
# ============================================================
# 2. IDENTIFIKASI PoT
# ============================================================

def is_pot_question(question):
    """
    PoT ditentukan berdasarkan adanya instruksi
    'Jawab menggunakan langkah perhitungannya'
    pada pertanyaan.
    """

    question = str(question).strip()

    return POT_SUFFIX.lower() in question.lower()


# ============================================================
# 3. LEVENSHTEIN / STRING SIMILARITY
# ============================================================

def levenshtein_similarity(s1, s2):

    s1 = str(s1)
    s2 = str(s2)

    if not s1 and not s2:
        return 1.0

    if not s1 or not s2:
        return 0.0

    return SequenceMatcher(None, s1, s2).ratio()


# ============================================================
# 4. DETEKSI TIPE JAWABAN
# ============================================================

def is_year(answer_str, question_str):

    answer_str = str(answer_str).strip()
    question_str = str(question_str).lower()

    year_pattern = r"\b(19|20)\d{2}\b"

    return (
        re.search(year_pattern, answer_str) is not None
        and "tahun" in question_str
    )


def is_numeric(answer_str):

    try:
        normalized = normalize_numeric_format(
            answer_str
        )

        float(normalized)

        return True

    except (ValueError, TypeError):
        return False


def get_answer_type(model_answer, gt_answer, question):

    # Year diperiksa lebih dahulu
    if is_year(gt_answer, question):
        return "year"

    # Numeric
    if is_numeric(gt_answer):
        return "numeric"

    # Selain itu text
    return "text"


# ============================================================
# 5. RELAXED ACCURACY
# ============================================================
def compute_relaxed_accuracy(
    model_val,
    gt_val,
    tolerance=0.05
):

    try:

        model_val = normalize_numeric_format(
            model_val
        )

        gt_val = normalize_numeric_format(
            gt_val
        )

        model_num = float(
            str(model_val).strip()
        )

        gt_num = float(
            str(gt_val).strip()
        )

        if gt_num == 0:

            return 1.0 if model_num == 0 else 0.0

        error_rate = (
            abs(model_num - gt_num)
            / abs(gt_num)
        )

        return 1.0 if error_rate <= tolerance else 0.0

    except (ValueError, TypeError):

        return 0.0

# ============================================================
# 6. EXACT MATCH
# ============================================================

def compute_em(model_answer, gt_answer):

    model_str = str(model_answer).strip()
    gt_str = str(gt_answer).strip()

    return 1.0 if model_str == gt_str else 0.0


# ============================================================
# 7. ANLS
# ============================================================

def compute_anls(
    model_answer,
    gt_answer,
    threshold=0.5
):

    model_str = str(model_answer).strip()
    gt_str = str(gt_answer).strip()

    similarity = levenshtein_similarity(
        model_str,
        gt_str
    )

    if similarity >= threshold:
        return similarity

    return 0.0


# ============================================================
# 8. NORMALISASI BOOLEAN
# ============================================================

def normalize_boolean_answer(answer):

    answer = str(answer).strip()

    mapping = {

        "True": "Ya",
        "False": "Tidak",

        "true": "Ya",
        "false": "Tidak",

        "TRUE": "Ya",
        "FALSE": "Tidak",

        "Yes": "Ya",
        "No": "Tidak",

        "yes": "Ya",
        "no": "Tidak",

        "YES": "Ya",
        "NO": "Tidak"
    }

    return mapping.get(answer, answer)

def normalize_closed_answer(answer):
    """
    Mengubah jawaban GPT yang masih berbentuk kalimat menjadi
    closed answer.

    Rules:
    1. Jika diawali Ya/Tidak -> ambil Ya/Tidak.
    2. Jika terdapat teks dalam tanda petik dua -> ambil isi petik.
    3. Jika terdapat pola 'adalah ...' -> ambil jawaban setelah 'adalah'
       dan buang keterangan setelahnya jika memungkinkan.
    4. Jika terdapat pola 'peringkat ...' -> ambil bagian setelahnya.
    5. Jika tidak memenuhi rule -> kembalikan jawaban asli.
    """

    answer = str(answer).strip()

    # --------------------------------------------------------
    # Bersihkan bullet/list di awal
    # --------------------------------------------------------

    answer = re.sub(
        r"^\s*[-•*]\s*",
        "",
        answer
    ).strip()

    # --------------------------------------------------------
    # Rule 1: Ya / Tidak
    # --------------------------------------------------------

    match = re.match(
        r"^(Ya|Tidak)\b",
        answer,
        flags=re.IGNORECASE
    )

    if match:
        return match.group(1).capitalize()

    # --------------------------------------------------------
    # Rule 2: Ambil isi tanda petik dua
    # --------------------------------------------------------

    quoted = re.findall(
        r'"([^"]+)"',
        answer
    )

    if quoted:
        return quoted[-1].strip()

    # --------------------------------------------------------
    # Rule 3: Pola "adalah ..."
    # --------------------------------------------------------

    match = re.search(
        r"\badalah\s+(.+)",
        answer,
        flags=re.IGNORECASE | re.DOTALL
    )

    if match:

        result = match.group(1).strip()

        # Kalau hasilnya memiliki koma/keterangan lanjutan,
        # coba ambil bagian pertama.
        result = re.split(
            r"\s+(?:dengan|yang|sebesar|yaitu)\s+",
            result,
            maxsplit=1,
            flags=re.IGNORECASE
        )[0]

        # Bersihkan tanda baca akhir
        result = result.strip(" .,:;")

        if result:
            return result

    # --------------------------------------------------------
    # Rule 4: Pola "peringkat ..."
    # --------------------------------------------------------

    match = re.search(
        r"\bperingkat(?:nya)?\s+(?:adalah\s+)?(.+)",
        answer,
        flags=re.IGNORECASE | re.DOTALL
    )

    if match:

        result = match.group(1).strip()

        result = re.split(
            r"\s+(?:dengan|yang|sebesar|yaitu)\s+",
            result,
            maxsplit=1,
            flags=re.IGNORECASE
        )[0]

        result = result.strip(" .,:;")

        if result:
            return result

    # --------------------------------------------------------
    # Tidak memenuhi rule
    # --------------------------------------------------------

    return answer
# ============================================================
# 9. RESOLVE PoT GPT
# ============================================================

def resolve_pot_answer(answer):

    """
    Mengubah script PoT menjadi final answer.

    Contoh:

    <comment>...</comment>
    <step>vals = [...]</step>
    <step>mean_val = np.mean(vals)</step>
    <step>Answer = mean_val</step>

    menjadi:

    2.1725
    """

    answer = str(answer).strip()

    # Hapus comment
    answer = re.sub(
        r"<comment>.*?</comment>",
        "",
        answer,
        flags=re.DOTALL
    )

    # Hapus tag step
    answer = answer.replace("<step>", "")
    answer = answer.replace("</step>", "\n")

    answer = answer.strip()

    # --------------------------------------------------------
    # Safe environment
    # --------------------------------------------------------

    safe_globals = {
        "__builtins__": {
            "all": all,
            "any": any,
            "len": len,
            "min": min,
            "max": max,
            "abs": abs,
            "round": round,
            "float": float,
            "int": int,
            "str": str,
            "sum": sum
        },

        "np": np
    }

    safe_locals = {}

    try:

        exec(
            answer,
            safe_globals,
            safe_locals
        )

        result = safe_locals.get(
            "Answer",
            None
        )

        if result is None:

            return None, "FAILED_NO_ANSWER"

        # numpy array
        if isinstance(result, np.ndarray):

            if result.size == 1:
                result = result.item()

            else:
                result = result.tolist()

        # numpy scalar
        elif isinstance(result, np.generic):

            result = result.item()

        return str(result).strip(), "SUCCESS"

    except Exception as e:

        return (
            None,
            f"FAILED_EXECUTION: {type(e).__name__}: {e}"
        )


# ============================================================
# 10. EVALUASI SATU SAMPLE
# ============================================================

def score_sample(
    model_answer,
    gt_answer,
    question
):

    answer_type = get_answer_type(
        model_answer,
        gt_answer,
        question
    )

    # --------------------------------------------------------
    # NUMERIC
    # --------------------------------------------------------

    if answer_type == "numeric":

        relaxed_acc = compute_relaxed_accuracy(
            model_answer,
            gt_answer
        )

        return {
            "score": relaxed_acc,
            "answer_type": "numeric",
            "metric_used": "Relaxed Accuracy",
            "em": None,
            "anls": None,
            "relaxed_acc": relaxed_acc
        }

    # --------------------------------------------------------
    # YEAR
    # --------------------------------------------------------

    elif answer_type == "year":

        em = compute_em(
            model_answer,
            gt_answer
        )

        return {
            "score": em,
            "answer_type": "year",
            "metric_used": "EM",
            "em": em,
            "anls": None,
            "relaxed_acc": None
        }

    # --------------------------------------------------------
    # TEXT
    # --------------------------------------------------------

    else:

        em = compute_em(
            model_answer,
            gt_answer
        )

        anls = compute_anls(
            model_answer,
            gt_answer
        )

        if em == 1.0:

            score = 1.0
            metric_used = "EM"

        elif anls > 0:

            score = anls
            metric_used = "ANLS"

        else:

            score = 0.0
            metric_used = "None"

        return {
            "score": score,
            "answer_type": "text",
            "metric_used": metric_used,
            "em": em,
            "anls": anls,
            "relaxed_acc": None
        }


# ============================================================
# 11. EKSTRAK ID DARI IMAGE
# ============================================================

def extract_id_from_image(image):

    """
    Mengubah nama image menjadi ID JSONL.

    Contoh:

    B115BARAverage_1.png
        ->
    B115BARAverage_1
    """

    image = str(image).strip()

    # Ambil basename
    image = os.path.basename(image)

    # Hilangkan extension
    image_id = os.path.splitext(image)[0]

    return image_id

def normalize_numeric_format(answer):
    """
    Menyamakan format desimal.
    
    Contoh:
    2,1725 -> 2.1725
    1,23   -> 1.23
    -3,28  -> -3.28
    
    Tidak mengubah angka yang sudah menggunakan titik.
    """

    answer = str(answer).strip()

    # Jika koma digunakan sebagai desimal
    if re.fullmatch(r"-?\d+,\d+", answer):
        answer = answer.replace(",", ".")

    return answer
# ============================================================
# 12. MAIN EVALUATION
# ============================================================

def evaluate_gpt():

    # --------------------------------------------------------
    # Load CSV
    # --------------------------------------------------------

    df = pd.read_csv(
        CSV_PATH
    )

    print("=" * 80)
    print("GPT-4o-mini EVALUATION")
    print("=" * 80)

    print(f"Total CSV records : {len(df)}")

    image_mapping = load_hf_image_mapping()

    # --------------------------------------------------------
    # Load JSONL GT
    # --------------------------------------------------------

    pot_gt = load_pot_ground_truth(
        JSONL_DIR
    )

    results = []

    # --------------------------------------------------------
    # Process every sample
    # --------------------------------------------------------

    for idx, row in df.iterrows():

        image = row["image"]
        question = str(row["question"]).strip()

        csv_gt = row["ground_truth"]

        gpt_raw = row["gpt4o_mini_answer"]

        # ----------------------------------------------------
        # Tentukan PoT
        # ----------------------------------------------------

        pot = is_pot_question(question)

        # ----------------------------------------------------
        # ID untuk mencari JSONL
        # ----------------------------------------------------

        image_id = extract_id_from_image(
            image
        )
        
        local_image_path = resolve_local_image_path(
            image_id,
            image_mapping
        )
        # ----------------------------------------------------
        # PoT
        # ----------------------------------------------------
        if pot:

            # GT berasal dari JSONL
            jsonl_record = pot_gt.get(image_id)

            if jsonl_record is None:

                print(
                    f"WARNING: JSONL GT tidak ditemukan "
                    f"untuk {image_id}"
                )

                gt_answer = None

            else:

                gt_answer = jsonl_record["final_gt_answer"]

            # GPT masih berupa script
            resolved_answer, resolve_status = (
                resolve_pot_answer(gpt_raw)
            )

            if resolved_answer is None:
                resolved_answer = ""

            # PoT hasil resolve sudah berupa final answer,
            # jadi TIDAK perlu normalize_closed_answer()
            # karena bisa merusak angka/jawaban yang valid.

        # ----------------------------------------------------
        # Non-PoT
        # ----------------------------------------------------

        else:

            # GT langsung dari CSV
            gt_answer = csv_gt

            # GPT sudah final
            resolved_answer = gpt_raw

            resolve_status = "NOT_POT"

            # Ubah jawaban GPT yang masih berupa kalimat
            # menjadi closed answer
            resolved_answer = normalize_closed_answer(
                resolved_answer
            )

        # ----------------------------------------------------
        # Normalisasi format numeric
        # Berlaku untuk PoT DAN Non-PoT
        # ----------------------------------------------------

        resolved_answer = normalize_numeric_format(
            resolved_answer
        )

        if gt_answer is not None:

            gt_answer = normalize_numeric_format(
                gt_answer
            )

        # ----------------------------------------------------
        # Normalisasi boolean
        # ----------------------------------------------------

        resolved_answer = normalize_boolean_answer(
            resolved_answer
        )

        if gt_answer is not None:

            gt_answer = normalize_boolean_answer(
                gt_answer
            )
            
        # ----------------------------------------------------
        # Score
        # ----------------------------------------------------

        if gt_answer is None:

            metrics = {
                "score": np.nan,
                "answer_type": None,
                "metric_used": "GT_NOT_FOUND",
                "em": None,
                "anls": None,
                "relaxed_acc": None
            }

        else:

            metrics = score_sample(
                resolved_answer,
                gt_answer,
                question
            )

        # ----------------------------------------------------
        # Simpan result
        # ----------------------------------------------------

        results.append({

            "image": image,

            "image_id": image_id,
            "image_path": local_image_path,
            "question": question,

            "is_pot": pot,

            "raw_model_answer": gpt_raw,

            "resolved_model_answer": resolved_answer,

            "ground_truth": gt_answer,

            "resolve_status": resolve_status,

            "answer_type": metrics[
                "answer_type"
            ],

            "metric_used": metrics[
                "metric_used"
            ],

            "score": metrics[
                "score"
            ],

            "relaxed_acc": metrics[
                "relaxed_acc"
            ],

            "em": metrics[
                "em"
            ],

            "anls": metrics[
                "anls"
            ]
        })

    # --------------------------------------------------------
    # DataFrame
    # --------------------------------------------------------

    result_df = pd.DataFrame(
        results
    )

    # --------------------------------------------------------
    # Save detailed result
    # --------------------------------------------------------

    # ========================================================
    # SAVE TO XLSX + EMBED IMAGE
    # ========================================================
    
    result_df.to_excel(
        OUTPUT_PATH,
        index=False
    )
    
    # --------------------------------------------------------
    # Buka workbook
    # --------------------------------------------------------
    
    wb = load_workbook(OUTPUT_PATH)
    ws = wb.active
    
    # --------------------------------------------------------
    # Buat kolom Image di paling kiri
    # --------------------------------------------------------
    
    ws.insert_cols(1)
    ws.cell(row=1, column=1).value = "Image"
    
    # --------------------------------------------------------
    # Masukkan gambar ke setiap row
    # --------------------------------------------------------
    
    TEMP_IMAGE_DIR = "/kaggle/working/compressed_images"
    os.makedirs(TEMP_IMAGE_DIR, exist_ok=True)
    
    for excel_row in range(2, len(result_df) + 2):
    
        image_path = result_df.iloc[
            excel_row - 2
        ]["image_path"]
    
        if (
            image_path is None
            or not os.path.exists(image_path)
        ):
            continue
    
        try:
    
            # Buka gambar asli
            img = Image.open(image_path)
    
            # Resize maksimal 600 x 400
            img.thumbnail(
                (600, 400),
                Image.Resampling.LANCZOS
            )
    
            # Path gambar hasil kompresi
            compressed_path = os.path.join(
                TEMP_IMAGE_DIR,
                f"image_{excel_row}.jpg"
            )
    
            # Convert ke RGB untuk JPEG
            if img.mode in ("RGBA", "LA", "P"):
                img = img.convert("RGB")
            else:
                img = img.convert("RGB")
    
            # Simpan JPEG terkompresi
            img.save(
                compressed_path,
                format="JPEG",
                quality=70,
                optimize=True
            )
    
            # Masukkan gambar hasil kompresi ke Excel
            excel_img = XLImage(
                compressed_path
            )
    
            excel_img.width = 300
            excel_img.height = 200
    
            ws.add_image(
                excel_img,
                f"A{excel_row}"
            )
    
            ws.row_dimensions[
                excel_row
            ].height = 155
    
        except Exception as e:
    
            print(
                f"ERROR embedding image "
                f"row {excel_row}: {e}"
            )
    # --------------------------------------------------------
    # Lebar kolom
    # --------------------------------------------------------
    
    ws.column_dimensions["A"].width = 45
    
    for col in range(2, ws.max_column + 1):
    
        column_letter = get_column_letter(col)
    
        if column_letter == "B":
            ws.column_dimensions[
                column_letter
            ].width = 25
    
        elif column_letter == "C":
            ws.column_dimensions[
                column_letter
            ].width = 70
    
        else:
            ws.column_dimensions[
                column_letter
            ].width = 25
    
    # --------------------------------------------------------
    # Freeze header
    # --------------------------------------------------------
    
    ws.freeze_panes = "A2"
    
    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------
    
    wb.save(OUTPUT_PATH)
    
    print(
        f"\n✓ XLSX saved to:\n{OUTPUT_PATH}"
    )

    # ========================================================
    # SUMMARY
    # ========================================================

    print("\n" + "=" * 80)
    print("SUMMARY")
    print("=" * 80)

    print(
        f"Total samples : {len(result_df)}"
    )

    print(
        f"PoT samples   : {result_df['is_pot'].sum()}"
    )

    print(
        f"Non-PoT       : {(~result_df['is_pot']).sum()}"
    )

    # --------------------------------------------------------
    # Overall score
    # --------------------------------------------------------

    valid_scores = result_df[
        result_df["score"].notna()
    ]["score"]

    print(
        f"\nOverall Score : {valid_scores.mean():.4f}"
    )

    # --------------------------------------------------------
    # PoT score
    # --------------------------------------------------------

    pot_scores = result_df[
        result_df["is_pot"] &
        result_df["score"].notna()
    ]["score"]

    nonpot_scores = result_df[
        ~result_df["is_pot"] &
        result_df["score"].notna()
    ]["score"]

    print(
        f"PoT Score     : {pot_scores.mean():.4f}"
    )

    print(
        f"Non-PoT Score : {nonpot_scores.mean():.4f}"
    )

    # --------------------------------------------------------
    # Breakdown answer type
    # --------------------------------------------------------

    print("\n===== ANSWER TYPE =====")

    print(
        result_df[
            "answer_type"
        ].value_counts()
    )

    # --------------------------------------------------------
    # Breakdown metric
    # --------------------------------------------------------

    print("\n===== METRIC =====")

    print(
        result_df[
            "metric_used"
        ].value_counts()
    )

    # --------------------------------------------------------
    # Failed PoT
    # --------------------------------------------------------

    failed_pot = result_df[
        result_df["is_pot"] &
        (
            result_df[
                "resolve_status"
            ] != "SUCCESS"
        )
    ]

    print(
        f"\nFailed PoT resolution : "
        f"{len(failed_pot)}"
    )

    if len(failed_pot) > 0:

        print(
            failed_pot[
                [
                    "image",
                    "resolve_status",
                    "raw_model_answer"
                ]
            ].to_string(index=False)
        )

    print(
        f"\n✓ Detailed results saved to:"
        f"\n{OUTPUT_PATH}"
    )

    return result_df


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    result_df = evaluate_gpt()
```
